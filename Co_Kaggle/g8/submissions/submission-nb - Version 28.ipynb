```python
"""Version 23: hierarchical MIMO HVAC with shared AHU and water coordination.

A bounded surrogate jointly scores shared AHU supply air and all connected
zone dampers, then coordinates reheat across the shared water plant.
This is not a calibrated plant model or a performance guarantee.
Runtime: O(AHUs * supply_candidates * zones * projection_iterations).
"""
import re
import numpy as np
from kaggle_evaluation import smart_buildings_inference_server


def project_shared_capacity(requested, priority, capacity):
    """Weighted Euclidean projection onto a shared nonnegative resource cap.

    Minimize sum_i 0.5 * priority_i * (x_i - requested_i)^2
    subject to 0 <= x_i <= requested_i and sum(x_i) <= capacity.
    Bisection on the common shadow price avoids SciPy and scales linearly.
    """
    requested = np.maximum(np.asarray(requested, dtype=np.float64), 0.0)
    priority = np.asarray(priority, dtype=np.float64)
    if requested.ndim != 1 or priority.shape != requested.shape:
        raise ValueError("resource vectors must have identical 1D shape")
    if (not np.isfinite(requested).all() or not np.isfinite(priority).all()
            or np.any(priority <= 0.0) or not np.isfinite(capacity)
            or capacity < 0.0):
        raise ValueError("invalid resource projection input")
    if requested.size == 0 or capacity == 0.0:
        return np.zeros_like(requested)
    if float(np.sum(requested)) <= capacity:
        return requested.copy()

    lower, upper = 0.0, float(np.max(priority * requested))
    for _ in range(32):
        price = 0.5 * (lower + upper)
        proposal = np.maximum(0.0, requested - price / priority)
        if float(np.sum(proposal)) > capacity:
            lower = price
        else:
            upper = price
    return np.maximum(0.0, requested - upper / priority)


class Version23Policy(smart_buildings_inference_server.Policy):
    """Hierarchical coupled-zone dispatch with explicit shared-equipment constraints.

    Action contract: continuous float32 in [-1, 1], discrete int32 in {0, 1}.
    All equipment topology and sensor locations come from episode metadata.
    """

    # Physical constants from the starter-kit action descriptions.
    MIN_SUPPLY_K = 285.0
    MAX_SUPPLY_K = 305.0
    MAX_RETURN_FRACTION = 0.70
    ACTIVE_THRESHOLD_K = 0.18
    MIN_VAV_FRACTION = 0.10
    MIN_FAN_PRESSURE_PA = 200.0
    # A cooling-only AHU admits at least 30% outside air.
    MIN_ECONOMIZER_OUTDOOR_K = 287.0
    MIN_ECONOMIZER_RECIRC = 0.15

    # One-step, deliberately uncalibrated zone thermal surrogate.
    # delta_temperature_K = AIR_GAIN * damper * (supply_K - zone_K)
    #                       + REHEAT_GAIN * damper * reheat_valve
    AIR_GAIN = 0.19
    REHEAT_GAIN = 3.6
    COMFORT_RESPONSE = 0.52
    SUPPLY_CANDIDATES_K = (285.0, 289.0, 293.0, 297.0, 301.0, 305.0)

    # Command-space budgets, NOT measured CFM/kW capacities.
    AIR_CAPACITY_FIRST = 0.80
    AIR_CAPACITY_ADDITIONAL = 0.50
    WATER_CAPACITY_FIRST = 0.80
    WATER_CAPACITY_ADDITIONAL = 0.42

    # Surrogate objective weights; tune only against held-out buildings.
    COIL_COST = 0.022
    FAN_COST = 0.080
    REHEAT_COST = 0.22
    ECONOMIZER_COST = 0.15
    WARM_SUPPLY_TIE_BREAK = 0.00001

    def __init__(self, building_meta_data, action_spec):
        """Index observations and actions once, never during each time step."""
        super().__init__(building_meta_data, action_spec)
        meta = building_meta_data
        self.n_continuous = int(meta["n_continuous_action"])
        self.n_discrete = int(meta["n_discrete_action"])
        self.fields = {str(s): i for i, s in enumerate(meta["observation_field_names"])}
        self.supply_indices = list(map(int, meta["temp_indices"]))
        self.pressure_indices = list(map(int, meta["pressure_indices"]))
        self.recirc_indices = list(map(int, meta["recirculation_indices"]))
        self.water_indices = list(map(int, meta["water_indices"]))
        self.ahu_count = len(self.supply_indices)
        self.water_supply = self.water_indices[0] if self.water_indices else None
        self.water_pressure = (self.pressure_indices[self.ahu_count]
                               if len(self.pressure_indices) > self.ahu_count else None)
        # The optional boiler switch follows all AHU switches. Never reuse
        # an AHU switch as a boiler when no hot-water plant is described.
        self.water_run = (
            self.n_discrete - 1
            if self.water_supply is not None and self.n_discrete > self.ahu_count
            else None
        )

        self.comfort = tuple(map(float, meta["comfort_temp_window"]))
        self.eco = tuple(map(float, meta["eco_temp_window"]))
        self.morning = float(meta["morning_start_hour"])
        self.evening = float(meta["evening_start_hour"])
        normalizer = meta.get("observation_normalization_metadata", {}).get(
            "zone_air_temperature_sensor", {})
        self.zone_mean = float(normalizer.get("sample_mean", 0.0))
        self.zone_std = float(normalizer.get("sample_std", 1.0))
        if not np.isfinite(self.zone_std) or self.zone_std <= 0:
            self.zone_std = 1.0

        def unit_order(key):
            match = re.search(r"\d+", str(key))
            return (int(match.group()) if match else -1, str(key))

        self.ahus = []
        assignments = meta["ahu_zone_assignments"]
        indices = meta["zone_action_indices"]
        for i, identifier in enumerate(sorted(assignments, key=unit_order)[:self.ahu_count]):
            zones = []
            for room in assignments[identifier]:
                vav = str(room).replace("room_", "VAV_")
                temp = self.fields.get(vav + "_zone_air_temperature_sensor")
                if temp is None:
                    continue
                actuators = indices.get(vav, {})
                zones.append({
                    "temperature": temp,
                    "low": self.fields.get(vav + "_zone_air_heating_temperature_setpoint_sensor"),
                    "high": self.fields.get(vav + "_zone_air_cooling_temperature_setpoint_sensor"),
                    "damper": actuators.get("supply_air_damper_percentage_command"),
                    "valve": actuators.get("reheat_valve_percentage_command"),
                })
            self.ahus.append({
                "zones": zones,
                "run": i,
                "supply": self.supply_indices[i],
                "pressure": (self.pressure_indices[i]
                             if i < len(self.pressure_indices) else None),
                "recirc": (self.recirc_indices[i]
                           if i < len(self.recirc_indices) else None),
                "outdoor": self.fields.get(
                    f"ahs_{identifier}_outside_air_temperature_sensor"),
            })
        # A shared boiler serves all mapped zones. Its rated capacity proxy
        # must not vary just because a different number of zones request heat.
        self.zone_count = sum(len(ahu["zones"]) for ahu in self.ahus)
        outside_stats = meta.get("observation_normalization_metadata", {}).get(
            "outside_air_temperature_sensor", {})
        self.outside_mean = float(outside_stats.get("sample_mean", np.nan))
        self.outside_std = float(outside_stats.get("sample_std", np.nan))
        self.reset()

    def reset(self):
        """No episode state: decisions depend only on the current observation."""
        return None

    @staticmethod
    def _clamp(value, low, high):
        return float(np.clip(value, low, high))

    @classmethod
    def _fraction_action(cls, fraction):
        """Convert physical 0..1 actuator opening to normalized action."""
        return 2.0 * cls._clamp(fraction, 0.0, 1.0) - 1.0

    @staticmethod
    def _temperature_action(kelvin):
        """Supply air 285..305 K maps to -1..1."""
        return float(np.clip((kelvin - 295.0) / 10.0, -1.0, 1.0))

    def _value(self, obs, index):
        """Read an optional sensor without indexing outside the observation."""
        if index is None or not 0 <= index < len(obs):
            return float("nan")
        return float(obs[index])

    def _schedule_fallback(self, obs):
        """Fall back to the occupied/eco schedule if zone setpoints are absent."""
        def clock(sin_name, cos_name, period):
            sine = self._value(obs, self.fields.get(sin_name))
            cosine = self._value(obs, self.fields.get(cos_name))
            if not (np.isfinite(sine) and np.isfinite(cosine)):
                return None
            return float(np.arctan2(sine, cosine) % (2 * np.pi)) * period / (2 * np.pi)

        hour = clock("hod_sin_000", "hod_cos_000", 24.0)
        day = clock("dow_sin_000", "dow_cos_000", 7.0)
        weekday = day is not None and int(day) < 5
        occupied = (weekday and hour is not None
                    and self.morning <= hour < self.evening)
        return self.comfort if occupied else self.eco

    def _read_zone(self, obs, zone, fallback):
        """Decode one zone (Kelvin); return None for an invalid temperature.

        The temperature observation is standardized. Setpoint observations
        are already in Kelvin and take precedence over schedule fallback.
        """
        raw = self._value(obs, zone["temperature"])
        temperature = raw * self.zone_std + self.zone_mean
        if not np.isfinite(temperature) or not 240.0 <= temperature <= 340.0:
            return None
        low = self._value(obs, zone["low"])
        high = self._value(obs, zone["high"])
        if not (np.isfinite(low) and np.isfinite(high)
                and 245.0 <= low < high <= 335.0 and high - low <= 35.0):
            low, high = fallback
        return {
            "actions": zone,
            "temperature_k": temperature,
            "cooling_error_k": max(0.0, temperature - high),
            "heating_error_k": max(0.0, low - temperature),
        }

    def _make_idle_actions(self):
        """A zero action is half-open, so explicitly set idle commands."""
        continuous = np.full(self.n_continuous, -1.0, dtype=np.float32)
        discrete = np.zeros(self.n_discrete, dtype=np.int32)
        for index in self.supply_indices:
            continuous[index] = 1.0  # Warmest: cooling coil effectively idle.
        for index in self.recirc_indices:
            continuous[index] = 1.0  # Max return air, physically capped at 70%.
        return continuous, discrete


    def _outdoor_temperature(self, obs, index):
        """Read outdoor Kelvin: the gateway normally standardizes this field."""
        raw = self._value(obs, index)
        if not np.isfinite(raw):
            return float("nan")
        if (np.isfinite(self.outside_mean) and np.isfinite(self.outside_std)
                and self.outside_std > 0):
            kelvin = raw * self.outside_std + self.outside_mean
        else:
            kelvin = raw
        return float(kelvin) if 210.0 <= kelvin <= 330.0 else float("nan")

    def _mixed_supply(self, coil_setpoint_k, temps, outdoor_k):
        """Model AHU common supply: recirculation, mixing, cooling-only coil.

        Return (recirculation fraction, mixed air K, delivered air K). The
        economizer is used only when it can eliminate coil cooling entirely;
        its extra exhaust-fan work is penalized in the planner.
        """
        return_k = float(np.mean(temps))
        outside_k = outdoor_k if np.isfinite(outdoor_k) else return_k
        recirc = self.MAX_RETURN_FRACTION
        if (self.MIN_ECONOMIZER_OUTDOOR_K <= outside_k < return_k - 1.5
                and coil_setpoint_k < return_k):
            min_mix = self.MIN_ECONOMIZER_RECIRC * return_k + (
                1.0 - self.MIN_ECONOMIZER_RECIRC) * outside_k
            if min_mix <= coil_setpoint_k:
                recirc = self._clamp(
                    (coil_setpoint_k - outside_k) / (return_k - outside_k),
                    self.MIN_ECONOMIZER_RECIRC, self.MAX_RETURN_FRACTION)
        mixed_k = recirc * return_k + (1.0 - recirc) * outside_k
        return recirc, mixed_k, min(mixed_k, coil_setpoint_k)

    def _zone_demand_model(self, states):
        """Prepare desired zone corrections and priority once per AHU.

        Positive delta-T means heating; negative means cooling. The model
        never requests reheat when the building has no usable boiler.
        """
        temperature_k = np.array([s["temperature_k"] for s in states])
        cooling_error_k = np.array([s["cooling_error_k"] for s in states])
        heating_error_k = np.array([s["heating_error_k"] for s in states])
        has_damper = np.array([
            s["actions"]["damper"] is not None for s in states])
        has_reheat = np.array([
            s["actions"]["valve"] is not None for s in states])
        can_heat = has_damper & has_reheat & (self.water_run is not None)

        needs_cooling = (cooling_error_k >= self.ACTIVE_THRESHOLD_K) & has_damper
        needs_heating = (heating_error_k >= self.ACTIVE_THRESHOLD_K) & can_heat
        if not np.any(needs_cooling | needs_heating):
            return None

        desired_delta_k = np.where(
            needs_cooling,
            -np.minimum(1.8, self.COMFORT_RESPONSE * cooling_error_k),
            np.where(
                needs_heating,
                np.minimum(1.8, self.COMFORT_RESPONSE * heating_error_k),
                0.0,
            ),
        )
        worst_error_k = np.maximum(cooling_error_k, heating_error_k)
        priority = (
            1.0 + 2.2 * np.minimum(worst_error_k, 5.0)
            + 2.0 * np.maximum(worst_error_k - 5.0, 0.0) ** 2
        )
        return {
            "temperature_k": temperature_k,
            "desired_delta_k": desired_delta_k,
            "needs_cooling": needs_cooling,
            "needs_heating": needs_heating,
            "can_heat": can_heat,
            "priority": priority,
        }

    def _evaluate_supply_candidate(self, model, coil_k, outdoor_k):
        """Estimate all zone effects under *one common AHU supply setting*.

        Airflow allocation is joint: increasing one VAV consumes a fraction
        of the shared fan capacity. The supply-temperature choice is joint:
        every zone sees the same delivered air, including neutral zones.
        """
        room_k = model["temperature_k"]
        desired_delta_k = model["desired_delta_k"]
        recirc, mixed_k, supply_k = self._mixed_supply(
            coil_k, room_k, outdoor_k)

        # Cooling demand is supplied by cold air; heating demand by VAV
        # reheat. These are desired damper fractions before shared limits.
        cooling_gain = self.AIR_GAIN * np.maximum(room_k - supply_k, 0.0)
        cooling_damper = np.where(
            model["needs_cooling"] & (cooling_gain > 1e-6),
            np.clip(
                -desired_delta_k / np.maximum(cooling_gain, 1e-6),
                self.MIN_VAV_FRACTION,
                0.98,
            ),
            self.MIN_VAV_FRACTION,
        )
        reheat_gain = self.REHEAT_GAIN + self.AIR_GAIN * (supply_k - room_k)
        heating_damper = np.where(
            model["needs_heating"],
            np.clip(
                desired_delta_k / np.maximum(reheat_gain, 0.15),
                0.24,
                0.88,
            ),
            self.MIN_VAV_FRACTION,
        )
        desired_damper = np.maximum(cooling_damper, heating_damper)
        air_budget = self.AIR_CAPACITY_FIRST + (
            len(room_k) - 1) * self.AIR_CAPACITY_ADDITIONAL
        extra_air = project_shared_capacity(
            desired_damper - self.MIN_VAV_FRACTION,
            model["priority"],
            air_budget,
        )
        damper = self.MIN_VAV_FRACTION + extra_air

        # Two inputs contribute to the same zone's thermal response:
        # cold/supply air and hot-water reheat.
        air_delta_k = self.AIR_GAIN * damper * (supply_k - room_k)
        reheat_valve = np.where(
            model["can_heat"] & (desired_delta_k > 0.0),
            np.clip(
                (desired_delta_k - air_delta_k) / (self.REHEAT_GAIN * damper),
                0.0,
                1.0,
            ),
            0.0,
        )
        reheat_load = damper * reheat_valve
        predicted_delta_k = air_delta_k + self.REHEAT_GAIN * reheat_load

        # Include spillover to neutral zones instead of optimizing only the
        # most overheated or underheated VAV.
        comfort_cost = float(np.sum(
            model["priority"] * (desired_delta_k - predicted_delta_k) ** 2))
        coil_proxy = max(0.0, mixed_k - supply_k) * float(np.sum(damper))
        fan_proxy = float(np.sum(extra_air)) ** 1.5
        water_proxy = float(np.sum(reheat_load))
        exhaust_proxy = (
            self.MAX_RETURN_FRACTION - recirc) * float(np.sum(damper))
        objective = (
            comfort_cost
            + self.COIL_COST * coil_proxy
            + self.FAN_COST * fan_proxy
            + self.REHEAT_COST * water_proxy
            + self.ECONOMIZER_COST * exhaust_proxy
            + self.WARM_SUPPLY_TIE_BREAK * (self.MAX_SUPPLY_K - coil_k)
        )
        return {
            "coil_k": coil_k,
            "recirc": recirc,
            "damper": damper,
            "valve": reheat_valve,
            "objective": objective,
            "water_load": water_proxy,
            "air_budget": air_budget,
            "predicted_change": predicted_delta_k,
            "requested_change": desired_delta_k,
        }

    def _ahu_candidates(self, states, outdoor_k):
        """Compute bounded SAT alternatives once; pricing does not re-simulate."""
        if not states:
            return []
        model = self._zone_demand_model(states)
        if model is None:
            return []
        return [
            self._evaluate_supply_candidate(model, coil_k, outdoor_k)
            for coil_k in self.SUPPLY_CANDIDATES_K
        ]

    def _plan_ahu(self, states, outdoor_k):
        """Single-AHU diagnostic plan, before cross-AHU water coordination."""
        candidates = self._ahu_candidates(states, outdoor_k)
        return min(candidates, key=lambda p: p["objective"]) if candidates else None

    def _coordinate_shared_water(self, candidate_sets):
        """Select AHU operating points using one building-wide water price.

        Without scarcity, each AHU keeps its independent best plan.
        Otherwise the boiler's shared capacity imposes a marginal price on
        the predicted reheat load. This can select a warmer common SAT and
        avoid promising all zones more hot water than the plant can deliver.

        The final valve projection remains necessary: finite candidate SATs
        and the surrogate model do not guarantee physical water feasibility.
        Pricing changes *upstream choices* while the projection enforces the
        bounded *command proxy* downstream.
        """
        def select_at_price(price):
            return [
                min(options, key=lambda p: p["objective"] + price * p["water_load"])
                if options else None
                for options in candidate_sets
            ]

        def load(plans):
            return sum(p["water_load"] for p in plans if p is not None)

        selected = select_at_price(0.0)
        capacity = self._water_capacity()
        if load(selected) <= capacity:
            return selected

        # Each candidate's score is affine in price. The selected aggregate
        # reheat demand is therefore non-increasing as price increases.
        low_price, high_price = 0.0, 1.0
        for _ in range(12):
            if load(select_at_price(high_price)) <= capacity:
                break
            high_price *= 2.0

        for _ in range(24):
            mid_price = 0.5 * (low_price + high_price)
            if load(select_at_price(mid_price)) > capacity:
                low_price = mid_price
            else:
                high_price = mid_price

        selected = select_at_price(high_price)
        if load(selected) > capacity:
            # Capacity may be infeasible even at the least-water setting.
            # Keep the least-water alternatives; the valve projection
            # provides the final safety bound.
            selected = [
                min(options, key=lambda p: (p["water_load"], p["objective"]))
                if options else None
                for options in candidate_sets
            ]
        return selected

    def _dispatch_ahu(self, ahu, states, plan, continuous, discrete):
        """Write the jointly optimized AHU/VAV commands; defer boiler control."""
        if plan is None:
            return []
        discrete[ahu["run"]] = 1
        continuous[ahu["supply"]] = self._temperature_action(plan["coil_k"])
        if ahu["recirc"] is not None:
            continuous[ahu["recirc"]] = self._fraction_action(plan["recirc"])

        dampers = plan["damper"]
        if ahu["pressure"] is not None:
            # Fan pressure responds to aggregate load AND a single highly
            # open branch; all VAVs share the same fan/static pressure.
            load_proxy = max(float(np.mean(dampers)),
                             0.50 * float(np.max(dampers)))
            pressure_pa = self.MIN_FAN_PRESSURE_PA + 1250.0 * np.sqrt(
                self._clamp(
                    (load_proxy - self.MIN_VAV_FRACTION) / 0.65, 0.0, 1.0))
            continuous[ahu["pressure"]] = self._clamp(
                pressure_pa / 1000.0 - 1.0, -1.0, 1.0)

        water_requests = []
        for state, damper, valve in zip(states, dampers, plan["valve"]):
            zone = state["actions"]
            if zone["damper"] is not None:
                continuous[zone["damper"]] = self._fraction_action(damper)
            if zone["valve"] is not None and valve > 1e-6:
                # All AHUs use the same hot-water plant. The actual valve
                # command is written after the building-wide allocation.
                water_requests.append((state, float(damper), float(valve)))
        return water_requests

    def _water_capacity(self):
        """Fixed building-wide reheat proxy, based on installed zone count."""
        return self.WATER_CAPACITY_FIRST + (
            max(0, self.zone_count - 1) * self.WATER_CAPACITY_ADDITIONAL)

    def _dispatch_water(self, requests, outdoors, continuous, discrete):
        """Project requests against one shared boiler capacity across AHUs."""
        if not requests or self.water_run is None:
            return
        demand = np.array([damper * valve for _, damper, valve in requests])
        priority = np.array([
            1.0 + 2.2 * min(state["heating_error_k"], 5.0)
            + 2.0 * max(state["heating_error_k"] - 5.0, 0.0)**2
            for state, _, _ in requests])
        budget = self._water_capacity()
        allocated = project_shared_capacity(demand, priority, budget)
        if not np.any(allocated > 1e-6):
            return
        discrete[self.water_run] = 1
        for (state, damper, _), load in zip(requests, allocated):
            continuous[state["actions"]["valve"]] = self._fraction_action(load / damper)

        coldest_error = max(
            state["heating_error_k"] for state, _, _ in requests)
        outside_k = float(np.mean(outdoors)) if outdoors else 294.0
        water_k = self._clamp(
            318.0 + 0.35 * max(0.0, 294.0 - outside_k)
            + 5.0 * coldest_error, 320.0, 349.0)
        if self.water_supply is not None:
            continuous[self.water_supply] = self._fraction_action(
                (water_k - 310.0) / 40.0)
        if self.water_pressure is not None:
            load_fraction = float(np.sum(allocated)) / max(budget, 1e-6)
            pressure_fraction = self._clamp(
                0.30 + 0.63 * load_fraction, 0.35, 0.98)
            continuous[self.water_pressure] = self._fraction_action(
                pressure_fraction)

    def get_action(self, obs):
        """Hierarchical decision: zone states -> shared AHUs -> shared boiler."""
        obs = np.asarray(obs, dtype=np.float64).reshape(-1)
        continuous, discrete = self._make_idle_actions()
        fallback = self._schedule_fallback(obs)
        # Phase 1: collect candidate operating points without issuing commands.
        # The central coordinator must see every AHU's predicted water demand.
        ahu_states, candidate_sets, outdoors = [], [], []
        for ahu in self.ahus:
            states = [self._read_zone(obs, zone, fallback)
                      for zone in ahu["zones"]]
            states = [state for state in states if state is not None]
            outside_k = self._outdoor_temperature(obs, ahu["outdoor"])
            if np.isfinite(outside_k):
                outdoors.append(outside_k)
            ahu_states.append(states)
            candidate_sets.append(self._ahu_candidates(states, outside_k))

        # Phase 2: price the shared boiler before committing AHU SAT choices.
        plans = self._coordinate_shared_water(candidate_sets)

        # Phase 3: write zone/AHU commands and enforce the shared water cap.
        all_water = []
        for ahu, states, plan in zip(self.ahus, ahu_states, plans):
            all_water.extend(self._dispatch_ahu(
                ahu, states, plan, continuous, discrete))
        self._dispatch_water(all_water, outdoors, continuous, discrete)
        np.nan_to_num(continuous, copy=False, nan=-1.0, posinf=1.0, neginf=-1.0)
        np.clip(continuous, -1.0, 1.0, out=continuous)
        return {"continuous_action": continuous, "discrete_action": discrete}

# ---------------------------------------------------------------------------
# Shared RL profiles and observation features (common/profiles.py)
# ---------------------------------------------------------------------------
# Each profile changes only V23's soft planning parameters, not its physical
# action bounds or device topology. Profile zero reproduces original V23.
PARAMETERS = (
    "ACTIVE_THRESHOLD_K", "COMFORT_RESPONSE", "AIR_CAPACITY_FIRST",
    "AIR_CAPACITY_ADDITIONAL", "WATER_CAPACITY_FIRST",
    "WATER_CAPACITY_ADDITIONAL", "COIL_COST", "FAN_COST", "REHEAT_COST",
)
PROFILES = (
    ("baseline", (1., 1., 1., 1., 1., 1., 1., 1., 1.)),
    ("comfort", (.5, 1.5, 1.25, 1.25, 1.25, 1.25, .6, .6, .6)),
    ("economy", (2., .7, .85, .85, .85, .85, 1.8, 1.8, 1.8)),
    ("air_capacity", (.7, 1.2, 1.5, 1.5, 1., 1., .8, .6, 1.)),
    ("heating", (.7, 1.3, 1.1, 1.1, 1.6, 1.6, 1., 1., .45)),
    ("cooling", (.7, 1.3, 1.3, 1.3, 1., 1., .35, .7, 1.4)),
    ("gentle", (1., .5, 1., 1., 1., 1., 1., 1., 1.)),
)
N_ACTIONS = len(PROFILES)
N_FEATURES = 18 + N_ACTIONS


class ProfileController:
    """V23 remains responsible for every 5-minute HVAC action."""

    def __init__(self, baseline):
        self.baseline = baseline
        self.original = {name: getattr(baseline, name) for name in PARAMETERS}
        self.profile = 0

    def select(self, profile: int):
        if not isinstance(profile, (int, np.integer)) or not 0 <= profile < N_ACTIONS:
            raise ValueError("Invalid profile")
        for name, factor in zip(PARAMETERS, PROFILES[profile][1]):
            # Always start from immutable original values, never compound.
            setattr(self.baseline, name, self.original[name] * factor)
        self.profile = profile

    def features(self, observation):
        """Only gateway observations/metadata; no simulator reward internals."""
        b = self.baseline
        obs = np.asarray(observation, dtype=np.float64).reshape(-1)
        fallback = b._schedule_fallback(obs)
        zones, outdoor = [], []
        for ahu in b.ahus:
            for zone in ahu["zones"]:
                state = b._read_zone(obs, zone, fallback)
                if state is not None:
                    zones.append(state)
            temperature = b._outdoor_temperature(obs, ahu["outdoor"])
            if np.isfinite(temperature):
                outdoor.append(temperature)
        heat = np.array([z["heating_error_k"] for z in zones] or [0.])
        cool = np.array([z["cooling_error_k"] for z in zones] or [0.])
        temp = np.array([z["temperature_k"] for z in zones] or [295.])
        clock = [b._value(obs, b.fields.get(name)) for name in
                 ("hod_sin_000", "hod_cos_000", "dow_sin_000", "dow_cos_000")]
        weather = np.mean(outdoor) if outdoor else 295.
        values = [
            *clock, (weather - 295.) / 30., float(bool(outdoor)),
            (temp.mean() - 295.) / 15., temp.std() / 10.,
            heat.mean() / 5., heat.max() / 10., np.mean(heat > 1.),
            cool.mean() / 5., cool.max() / 10., np.mean(cool > 1.),
            len(zones) / max(1, b.zone_count),
            float(b.water_run is not None),
            (fallback[0] - 295.) / 15., (fallback[1] - 295.) / 15.,
            *np.eye(N_ACTIONS, dtype=np.float32)[self.profile],
        ]
        features = np.asarray(values, dtype=np.float32)
        if features.shape != (N_FEATURES,):
            raise ValueError(f"Unexpected feature shape: {features.shape}")
        return np.clip(np.nan_to_num(features, nan=0., posinf=5., neginf=-5.), -5., 5.)

    def get_action(self, observation):
        return self.baseline.get_action(observation)

# ---------------------------------------------------------------------------
# P3 network (p3/model.py)
# ---------------------------------------------------------------------------
import torch
from torch import nn


MARGIN = 0.03


class QPolicy(nn.Module):
    """Predict normalized discounted reward; retain V23 on small advantages."""

    def __init__(self):
        super().__init__()
        self.network = nn.Sequential(
            nn.Linear(N_FEATURES, 64), nn.ReLU(),
            nn.Linear(64, 64), nn.ReLU(),
            nn.Linear(64, N_ACTIONS),
        )
        nn.init.zeros_(self.network[-1].weight)
        nn.init.zeros_(self.network[-1].bias)
        # Included in snapshots sent to simulator workers.
        self.register_buffer("epsilon", torch.tensor(0.0))

    def q_values(self, features):
        return self.network(features)

    def greedy_action(self, features):
        scores = self.q_values(features)
        # V23 wins ties; only switch when the predicted advantage is positive
        # by more than the margin (in normalized reward units).
        adjusted = torch.cat((scores[..., :1], scores[..., 1:] - MARGIN), dim=-1)
        return adjusted.argmax(dim=-1)

    def forward(self, features):
        """Adapt epsilon-greedy behavior to common.rollout's categorical API."""
        action = self.greedy_action(features)
        one_hot = torch.nn.functional.one_hot(action, N_ACTIONS).float()
        probs = (1 - self.epsilon) * one_hot + self.epsilon / N_ACTIONS
        logits = probs.clamp_min(torch.finfo(probs.dtype).tiny).log()
        return logits, torch.zeros_like(logits[..., 0])

    def validate(self):
        if not all(torch.isfinite(t).all() for t in self.state_dict().values()):
            raise ValueError("Nonfinite model state")
        if not 0 <= float(self.epsilon) < 1:
            raise ValueError("Invalid exploration rate")


def signature():
    return {"schema": 1, "algorithm": "baseline_gated_double_dqn",
            "features": N_FEATURES, "hidden": [64, 64],
            "margin": MARGIN, "baseline_version": 23,
            "profiles": [[name, list(factors)] for name, factors in PROFILES]}

# ---------------------------------------------------------------------------
# Embedded P3 checkpoint (exact Git blob, no downloads or Kaggle data attachment)
# ---------------------------------------------------------------------------
import base64
import hashlib
import io
import math

CHECKPOINT_GIT_SHA = "6925108033e5ff1f812f55ef78bf4d080ea60025"
CHECKPOINT_BASE64 = (
    "UEsDBAAACAgAAAAAAAAAAAAAAAAAAAAAAAANABUAYmVzdC9kYXRhLnBrbEZCEQBaWlpaWlpaWlpaWlpaWlpaWoACfXEAKFgF"
    "AAAAbW9kZWxxAX1xAihYBwAAAGVwc2lsb25xA2N0b3JjaC5fdXRpbHMKX3JlYnVpbGRfdGVuc29yX3YyCnEEKChYBwAAAHN0"
    "b3JhZ2VxBWN0b3JjaApGbG9hdFN0b3JhZ2UKcQZYAQAAADBxB1gDAAAAY3B1cQhLAXRxCVFLACkpiWNjb2xsZWN0aW9ucwpP"
    "cmRlcmVkRGljdApxCilScQt0cQxScQ1YEAAAAG5ldHdvcmsuMC53ZWlnaHRxDmgEKChoBWgGWAEAAAAxcQ9oCE1ABnRxEFFL"
    "AEtASxmGcRFLGUsBhnESiWgKKVJxE3RxFFJxFVgOAAAAbmV0d29yay4wLmJpYXNxFmgEKChoBWgGWAEAAAAycRdoCEtAdHEY"
    "UUsAS0CFcRlLAYVxGoloCilScRt0cRxScR1YEAAAAG5ldHdvcmsuMi53ZWlnaHRxHmgEKChoBWgGWAEAAAAzcR9oCE0AEHRx"
    "IFFLAEtAS0CGcSFLQEsBhnEiiWgKKVJxI3RxJFJxJVgOAAAAbmV0d29yay4yLmJpYXNxJmgEKChoBWgGWAEAAAA0cSdoCEtA"
    "dHEoUUsAS0CFcSlLAYVxKoloCilScSt0cSxScS1YEAAAAG5ldHdvcmsuNC53ZWlnaHRxLmgEKChoBWgGWAEAAAA1cS9oCE3A"
    "AXRxMFFLAEsHS0CGcTFLQEsBhnEyiWgKKVJxM3RxNFJxNVgOAAAAbmV0d29yay40LmJpYXNxNmgEKChoBWgGWAEAAAA2cTdo"
    "CEsHdHE4UUsASweFcTlLAYVxOoloCilScTt0cTxScT11WAkAAABzaWduYXR1cmVxPn1xPyhYBgAAAHNjaGVtYXFASwFYCQAA"
    "AGFsZ29yaXRobXFBWBkAAABiYXNlbGluZV9nYXRlZF9kb3VibGVfZHFucUJYCAAAAGZlYXR1cmVzcUNLGVgGAAAAaGlkZGVu"
    "cURdcUUoS0BLQGVYBgAAAG1hcmdpbnFGRz+euFHrhR64WBAAAABiYXNlbGluZV92ZXJzaW9ucUdLF1gIAAAAcHJvZmlsZXNx"
    "SF1xSShdcUooWAgAAABiYXNlbGluZXFLXXFMKEc/8AAAAAAAAEc/8AAAAAAAAEc/8AAAAAAAAEc/8AAAAAAAAEc/8AAAAAAA"
    "AEc/8AAAAAAAAEc/8AAAAAAAAEc/8AAAAAAAAEc/8AAAAAAAAGVlXXFNKFgHAAAAY29tZm9ydHFOXXFPKEc/4AAAAAAAAEc/"
    "+AAAAAAAAEc/9AAAAAAAAEc/9AAAAAAAAEc/9AAAAAAAAEc/9AAAAAAAAEc/4zMzMzMzM0c/4zMzMzMzM0c/4zMzMzMzM2Vl"
    "XXFQKFgHAAAAZWNvbm9teXFRXXFSKEdAAAAAAAAAAEc/5mZmZmZmZkc/6zMzMzMzM0c/6zMzMzMzM0c/6zMzMzMzM0c/6zMz"
    "MzMzM0c//MzMzMzMzUc//MzMzMzMzUc//MzMzMzMzWVlXXFTKFgMAAAAYWlyX2NhcGFjaXR5cVRdcVUoRz/mZmZmZmZmRz/z"
    "MzMzMzMzRz/4AAAAAAAARz/4AAAAAAAARz/wAAAAAAAARz/wAAAAAAAARz/pmZmZmZmaRz/jMzMzMzMzRz/wAAAAAAAAZWVd"
    "cVYoWAcAAABoZWF0aW5ncVddcVgoRz/mZmZmZmZmRz/0zMzMzMzNRz/xmZmZmZmaRz/xmZmZmZmaRz/5mZmZmZmaRz/5mZmZ"
    "mZmaRz/wAAAAAAAARz/wAAAAAAAARz/czMzMzMzNZWVdcVkoWAcAAABjb29saW5ncVpdcVsoRz/mZmZmZmZmRz/0zMzMzMzN"
    "Rz/0zMzMzMzNRz/0zMzMzMzNRz/wAAAAAAAARz/wAAAAAAAARz/WZmZmZmZmRz/mZmZmZmZmRz/2ZmZmZmZmZWVdcVwoWAYA"
    "AABnZW50bGVxXV1xXihHP/AAAAAAAABHP+AAAAAAAABHP/AAAAAAAABHP/AAAAAAAABHP/AAAAAAAABHP/AAAAAAAABHP/AA"
    "AAAAAABHP/AAAAAAAABHP/AAAAAAAABlZWV1WAgAAABpbnRlcnZhbHFfSwxYBgAAAHVwZGF0ZXFgSxRYDwAAAHZhbGlkYXRp"
    "b25fbG9zc3FhRz+L2hLn7LqNWBAAAAB2YWxpZGF0aW9uX3N0ZXBzcWJNYAN1LlBLBwh/c8kyZgYAAGYGAABQSwMEAAAICAAA"
    "AAAAAAAAAAAAAAAAAAAAABQAGABiZXN0Ly5mb3JtYXRfdmVyc2lvbkZCFABaWlpaWlpaWlpaWlpaWlpaWlpaWjFQSwcIt+/c"
    "gwEAAAABAAAAUEsDBAAACAgAAAAAAAAAAAAAAAAAAAAAAAAXADoAYmVzdC8uc3RvcmFnZV9hbGlnbm1lbnRGQjYAWlpaWlpa"
    "WlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaNjRQSwcIP3dx6QIAAAACAAAAUEsDBAAA"
    "CAgAAAAAAAAAAAAAAAAAAAAAAAAOAEIAYmVzdC9ieXRlb3JkZXJGQj4AWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpa"
    "WlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpsaXR0bGVQSwcIhT3jGQYAAAAGAAAAUEsDBAAACAgAAAAAAAAAAAAA"
    "AAAAAAAAAAALAEEAYmVzdC9kYXRhLzBGQj0AWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpa"
    "WlpaWlpaWlpaWlpaWlpaWgAAAABQSwcIHN9EIQQAAAAEAAAAUEsDBAAACAgAAAAAAAAAAAAAAAAAAAAAAAALAEMAYmVzdC9k"
    "YXRhLzFGQj8AWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpa"
    "4v9nPkq8Aj5ATV09Q0JfPlh9M72UTiU9oiuNvj7kez47NLA+VOqMPYrpqj65AuM9t0KsPjFiwj2GexI+y4OWPMMahj7r+Oc8"
    "0CFevQFaID36mIi9RlFQvcHXI7y0ggU+eQTlvSKUKL1s0LO9lilWvV6cLD3VRkq+7fQ4PmEdfb4iAXc+ilA6PngvBj4sgYc+"
    "c77/PUobqj5t7gI++nJTvUyBhT0H9kc9zqk9PZNzQz6OS9M90Mt2vfwr/j0qOBY9vt7ePeYF3L2N3oG+ME33vcDmwL246kg+"
    "jPRrPXKpqT3rLig+WOyrPU+rlz5s4To9C70UPuN6gjvag4o+2CSWPdQ3mD2Rgeq8CIGdPml7Jb6n16i9KPQivoSfIz5mMMQ9"
    "T8hBPmk6Eb5nj1C+eOSUvWYwNb5+5yc+senAPV8sKj4gh9O9GChPvvmWPT6TD5g9+cShPq8T0D3pK2A+ndqLPM6pTDwodJM9"
    "m1ZaPeseeT0/bJ49HLkPPvH5Fb4oioO9UTQ/PsxFgb0uiYG9EiInvtOQ6r2go0w98EHYvFWpFL7AuZk7JeULvmvALb4qjeG9"
    "Uj4zvmpnAr5LuEw+9LkaPcRxfD2FAz++3YEGvnZWiL1oGQA9ODU0vlB8sL3zNvW9AEsROtppmL2AEWO8UMcKvuCSDL6aeAy+"
    "zrkavizlq73QN0g+7JQivZE1MD4mkB4+dlY3vvY3Dz7eP929NYOJvWt6D75ZVHy+nAahPVGQX7z0/tw9PmY5vjfpMD4j5/m9"
    "S54Ru8yekz30T9g9ItEnPnVIYD7R5/u97i74Pce08L1QvsU9U7zlvTmLOj4lUBa+oMgRPThoOD5hiWY+RbrAPTLQqz5WfhU+"
    "QRNRPnxKgL3NmuA9qZ+Mu/sUjjyeIRI+lR4/PcPlpb2N00y9n0+ZO/BA7D3xlhG9q+7RvZQNY7te3EO+ZyAgPioeLj6XzkC+"
    "gEI+PZnAKz3nwNg9QryoPR3slz7Kf4E+6DY5PgpeCz4CAXo9l5ajPD2PFr27D1M9yXSfPSE9MT7aFSG9JVt1vIxyITs6xQq8"
    "h702Pc02gD2y8sq9bxEMvpejfz6NStE94PoivEDz9r0gCkO+c3TTvS6cuD3R94U+93uNPhwdTz4U4l+6SwEyu4rArLwS11A+"
    "MjwSPg9Nwr1iAbE9yei+vX6Irb23rxQ+6xqjvbviAb0c6D2+tmuIPN4NRD70dxc+QC/kvRMgSr6SVSa+iZWVPj4v1j2YV509"
    "xyLguwRgLL0XgzY9J74Uvhw7Ir58mhe+Gpb3PbtYTL5V98M8JWmxvXtqOjwCXhW+BtJpve7+1j3+3Yu9uKa8PQnsuzzUPUC+"
    "bAaPPde+mr2YoHq9/fdCvo3LQT21HHy9nv2BPv00kT6qH5Q+Y9YTvev4qD4czOY8jxKcPUNfQj4EAMo92ylivWMIU730ci6+"
    "j071PQFkP76cjQI+ZtwWPgbIRj6NcdA8t/8/voVBED5WFRA+2NZLvuzxOb3qJxK+qsqOPVPEHj0D/cU9Vx+LPS/du70YLak+"
    "Q7yLPj+pC7zHcpk4shyVPWLUXb6ifAq+qT7VPH+FnL0nRui9Nm2nvO9khb3wXJe93xuYPZJrnL1BWwE9Rct+vf6rJr7tuSs+"
    "pFESvo0LEj71U7s9K7thPnaVrj74g1Q+JthoPM+yTT5BEVU+ZhWRPeeBfD61bSQ+O+K3vANwvDxMPaG9sdWWvIDbGz4yK+I9"
    "4zrIvU8ijb2j25q+iaSbvg4rab0Ilkq9ATcKPpEyzr2ZZti928ghvtCPsLyG3wC+TTg4vqTTxr07VYG+qujDPQ7rzD2g5lO+"
    "LLkDPWTHsr2vNIq9n3irvIp8QT4iMRK93KfFvJdzHj4ZMwE+WoWIPQpcaD4XJgo+NQ9LvrBOkTzW1Dg9d5GCPSmgGT0QTsY+"
    "GDyYPY4mJLzQkay7khz3PeXbxz1BC1o9sOYVPlQhRb4bZS6+WhRhvK4zWD5Ur+49eEvjvZSls7xaEQ6+7sAXPCgT3bz6oC8+"
    "QAhWPjMHPL6x7gg+/s14vtQ2Mj40Bao+EIhVPs5DrT6ZtHy9jcDePa4Pcj5GMyq89xr+u+KdCj2faOY9g1/uPc4OUj29C669"
    "c5sKvucbUT1zfKg9qx2jvQzIDr7RR0Q+JldUvVqmVz0Im0m9+3Y5vrNHXr25Fai98Pdzvqoa771jokG+Du51vlFT0juWZNK9"
    "AjZJPgLAL70SWmq+rvknPolmNr0qVjy96LWdPICYSr3LMvu9WRoNPoLZGD5hTxu95+0lvmPrib60ZzW9fI5QPZojwT1R7oW9"
    "fz60ve9pH71gHba+/Ta1vPj4Xb7KR1O+wjo3vo8YvD2efuo7ruG7PEr8sD0wOjc+QZRePtwJsTxxv2G9x49APiYuYD4pIUq9"
    "03x+PZU8g77VFJS92XWDPYTvWT2opy09xrvUvVIWhT6f5SQ+ChtDPXtY7D2WpQo+10ZRPmX4Or3nMro9dpKwvez1l72NkxG+"
    "5HJNPrt0Jz6Jp2Q+eHv/PTNlHz7VWwO6CyWovQ45Kj6WQBm+26a5vHyHAr2s6is9VtzSPSrK/L2Dg0O9UVoQPg4o5DzGyay8"
    "CIAzPi2plD1Hi4E+6bhXPpaBmz399js+d5U+O6e5n7yuS4O8htZ4PW7ibz0PsDQ+JFcBvJ0CNr2oYUa9QOPFvbRLCz1l3xg+"
    "IyYgPs4OvD0CuBa+k9XbvdT8M72U/xa9MMGGPCUAPL7E+H49eDwcvgyWAb5jRCq+P3IyPuvuQb6Lfho+wE/cu0CLvrwXGQA+"
    "8G6XvPFwAD7PEhQ+y8H+veTyEz3lW4Y9cR6NPUAx7zycE2A9zlNyvmF4jz3hI68+TFStPjNZCz7s2b69YMRSPtcd4bzAnjE+"
    "e4HivX3K1T05mbG95bVYPb7+Jj45SBU9eFz5PYgkLT5JcSw+G0rKvQ7XXj0/q8k8dDkXPcaCFT4A+DK99ecKvpmngb3aixE+"
    "NJCCPl7P0j5Bq5Y+sHuNPkoSST6Dape969ytvTXzjL3ygF69I4nKvbMk770pVWi+nIPTvezED76XApm9vk2YPaGoLj6s3GE8"
    "/9HvveK+Oj5IIYu9F6zBvd5d/D1xnYW+x4HmPUiTrT7Zl3M+dvZXPp3E9Lt4aTc+Tm4FPq1hiT0NAMQ9e1WWPmxHRL7ighA8"
    "xCm7vbYhyr0/FBG8/AEOvkD1Dj5Zn6Q8mjidvcFYjT0XtiS+Cqk6vXSYH71jBy6+WyI4vsRaMD118U0+oayNPunxLz5tt6u9"
    "dRJtvaLQWz7bJdk9UUsmPYWmKT0IWBC+IfAzvZdkVj4FCVu9gZrbOjiuLL7HqKW9EPHuPEDuIL5NVEY7etZpPSES4b30T3M9"
    "WmoHvqGbHD0yA/k9ArXHu+bjgr660oC+6eKOvjz6jr7LMZy++HwoPQzJAj5Pea++bD2SPmFBLz7Ix869tRPdvC0ppD1Zs0Q9"
    "SethPSkXgryr+mQ+rNIsvlMSiD7yPyI9VK0kPWHFHz719bi9prgIvXxbAD6Hibc+pDxHO+0cJj5lRy+8hTRqPkkipT1oT3k8"
    "M3pVPvkER72sSOI9btsLvt+ERjy9kKE9NF1JPkDMzj34gwW+BTuAPndNyLyj69892TfsvSz0AL1L3BA+4iLvvf/9ijyQeF0+"
    "ALMSPg+cqD5HMbS9HBTqPVD4x72YEsy8pWvvPfflkz4Wzyw9wW1cPqmWmT1p/i0+nhsNPk98Pj5dwhq8ZaJMPkw/VD6HAxG+"
    "8IA9u73mEb5Mu009IAAmvvbtPjxyTRY+D1LEvDiWez6dLwI+A4RDPlkEkT6Gyru9wDVIPlokOD4X1zw+Y81jvqShfr3VIJC9"
    "NicYvEzS8z3dcZw9N7ExPmNngr0M6r09MGA7Pfla6L2aFCq+d4nUvQMyDD7i1VO+nnvxvd7qCD405Bc+yzs8PgxpAb7t2+m9"
    "NjL9PbV4Vj486R4+62kEPfiouT1601o+ce69va4WTL3v2i8+fv0Xvrltmr0gwDM+/BP3PRWEOz16SBs9W94JPY9wOz6j2fW9"
    "sIsbvvNNZD6mVhc9IwjXPj2siT1Yfro9t0olPsODWjwowiK+cK4ZPpKlKj4BIFG+AP/LvTYqMT5drc493rWvPd3Um70FK3y9"
    "WuQMvcX5TD3jlS2+B10IPgZimj2wdM+8aCLjPBbSv72KoSU+9l9vPh9EwT1pFTy9JCJhPlKVaD44crA89e12Pos6LL3sfoM+"
    "JobEPUdm0L2fP949CrgDvmY4pr15i+q8oAnQveWXTT6hvDG9OA5uPcto8L3yVw8+FJh+vRXJG75fv6A9CTctPonEZz4ptpE8"
    "ie+AvXWLBj5YGmA+46snPhp27D0MeSK9vhlFPkrt8D0ClFm9XDUBPkJO4T2CpzY+Ky2nvSxkwL3qv3K9KYEevTShgzxFAr69"
    "gF0ZPgTLkr1ycfc9YQ7zPeVSIL63yDe+oq+Jvh/lGz1btja+/DgKvs4WHbwYO1S9EqrsPbGXmb66LIy9ffQdvhpPgTzyqbW8"
    "X8+LPZpBZTxKZ/89o0VIPncHBD7QmqS9R1OovQ0eBj6H8Mq9jm/QPfEd1j1Yh1M+4lSyPnGpyz79G0k+TdbEvIjNvz5XHAQ9"
    "jCAhvR3UYz62OIQ+/ZZ6Pak0bD2qVA6+fqUGPYLhTD6GJpw9zfhoPVwwY7su6Cs942NVPS4oKT3sxxE+TP8QvUL1sj2XbgS+"
    "qqYXPtedjz7+lag+GtqqPs6oSj4h89Q9ma/PvQXbgD0GmMk8Kj2TPW5IW70ouyQ8Ls7hPYbrkT28Vji9GcMFvda8sjtf6MO9"
    "HRCMPuAUBz0zk4Q+SVVgPXQ1JT3qCI492z+dvgbRDT7B6yM+/7TZPVzYMzz4HI8+YGy1PvPZVT7662s+ongqvVhNrD0Wf1U9"
    "Ch4EvMKoebyxYem9zxQbPWflbr0MQiU8QORLPowpfL1AGoG98h33PTyXlr2Ys4y8yTo/Pt3sp70ahBK+5zrlvaAeY73NWsK9"
    "8I3UvaZ0pT2INgs95BiDvYgcH75094C9A96svaZKhb06wTi+wHowPa14Fr6Cfx++ElQnvtr1qj2d8CG9V3EFvk3bdz7xsAE+"
    "btzqPX11Qr6RtlS+F6uuuoqSoT4p1lU+Oiqhu8B21j2xHTc8EReKPUUHS71j2DY+XSwtPotRor3WpB89IQOSvR825j1k67E9"
    "pF5mPZGklb0vgTg+lRyTPfVGCr4DZEE+EA1BPgnPSz7ksXI9Bf8LvhRGNT5QNaE+U5ZdvDGy5z1fFy69KspOPju5Oz5KeMO8"
    "YhY+OqxZfD4vAY+9T5/rvSCo9D309I29/p0xvW07Sbs3dHS8wSQnvrXfRT7QWTG+YN/TOxGcmj0zwdW980y9vdGf0b19VwO8"
    "c2H8POJgnD1AriY+nZe5vQKQFb0Q+Kc9M3pJPWn/Bj47KJS90HolPaWf5z02HOg9K8L1vW9FjL3DQB2+963OveUR1j2eEre9"
    "kSaVPcxCJ77P8ec987/3vXrBzD3givk6epZGvAo6or6Fwzq+Yv5aPS2SwDzzwoy+ORuZvcFnpT3h4W69Mm52u55eBD4BYvc9"
    "+kASPSbNEzx5gsc9WFsFPhFp7D1kd3W8pAQ/PjEONb5FL9e9S8RNvQGcST4AgCE3dlJPPelhs7193YO94N9ZvqD0Lr6RvTU+"
    "xXIdPoq5hz4emfu9yNIEPg0Trr3OYeg9XZIWPkwDMj7W5Bs+T5kMPqgtsTr6gfw9iuYHvp9QGD4KyM+9e+AGPoAu2r00M1a9"
    "x2XXvU9uEr50z5m61w0CvBzxaD54ZJc+H5uePr94oj71Vuo8QlsOvlDgCT6rdWY+ic4nvrTV6z0r9zY+ifh5PSjVnb2J5jo+"
    "qfbZPfQMJz7jF0W++HLVPfpM2L2samI+I4ytvb7P9j2gzNo9szRWPRlzb72HsBA8EpLYvenfTL4JAyK+Te4UPlTxYbxfVtA9"
    "vX0jviMHVz5dJx4+RgEFPqRhfL2nn0O+4hLuOU22gb0FD0E+JO9KvnshXL1IiF8+noQKPjaqiz0yDEW+aOJIPsk/Sj0m7kO9"
    "mifFPr7jNz3VXgg+UajDPuMe1DxZ2vK9BBZjPQUgFj6zRsi7QToovllT4bxrqpe97ygmveAR+b2AML49g5FUPPSFY779VDg+"
    "lNvXPYPJwTy3ffC9ABT3PH+Dhz4548U90Gp3vl64QD2fhYS9IlovPUQJnTzoSOE9CXAjvguMLz4t7kW+HKoTPX5/BL7k8yw8"
    "/Hd4vK+t9byEPYG9RFEkvH7vYz4RnUs+Lk82PeobHz5x6WQ+oDgXvNhFgjy1Zom9VqZdvfKMAzynkXU+PDaqPSxO9T2noiM+"
    "RUogPpOdOz6+0Fi8KsyNPgPhFj6Ki9i9iE0Uvjn1CL5KheW9F7E0vX/MSDwwqPO8OkxfvAlLrL14yk48DBFWPkDPl70SoxK+"
    "YZsFPeeYmz3sfpo+bVyiPuVQpz6gbhq94EagPhN4kj75bsS9+yo/PnxBlz71hmi9JcbLPXYOyL1A6x++VvD+PJ3KlD1XHee9"
    "8KzKPVw6Qb4Mci09r+tAPtfGtz1WfwC+LlfHPVZHs7056SI+Ny5oPpewtj4CIoY9xH+fPQZGfz75ziE+fsI7Pivpqb1GygM+"
    "CR3yulkPWb3gDwY+Rz8hPcik570oC5S7VLiBven5k7sIEMk8GVx6vpZNID2abKW7gJBxvY8qPD5p7MO9wZhFPoVDPj53yCU7"
    "v/I1PlvZhz5Gw5c8l9WhvJh1Rz7IjCM+JsFkvETvmD0KH8K8v01QvhPcHj3G0Ik9PGd6vUpqdj0qhiW+8JLdvFRVej678wU+"
    "m2h/PL7mhj2+LhS+PbW/PV84IL1vEHC+H5iPvhPxJDxGIr89KRFdvqcGNzxGEpo9rUY8PuTtjzzNT2I+uJRavePuHz4/ws09"
    "YYntvfmjL76E90M9Aa6kPIU/1LysGf28jq6UPQP9lT0wGdy9xgfnPSPTkD3eT4i9vwsaPuwLQj5vjFM+IYiTPr2jZz7RAIQ9"
    "EkI+Pq05j73SnUE9a3L2vPRhDL6XDPU7N4vMvTfUyTyWDLQ86b3jPFIcK779h1S+MvXjvbkfJj5a7uO9praXPfjBFr6igqC9"
    "jHCaPhWBnD4DrEI98UGKPrLSVT537Eo+y6+fPuH1Nb348Dy8tPGfPZ2hFLyPkU89/dHkPF/Q0r0wa0u9lMDXPT+yGb6GT5O9"
    "+lSTPSC3971ykCE+RnMJPWnIDz5AgpG8yjBLvquHZT4EbDA+zhUDPpnWuztzGxU+ZzryvPoyJj4Q9OQ9pG8mvgqUAL3dEDi+"
    "KEY2PhJnHD7sWy891sRHPvRNhb2ZnSa+nIypPTYBPj480k4+qkEFO2g4Wj7qEfs9VToUPnnTvr4xrYw+V5m6vQOunT6j6oQ9"
    "eXw7vkWz4z2OTRS9z9bdvRDTm7psv9k84JVTPlbL77zvdbs945IzPXVGnjypmNg9Cem8vQ0uOb7IebQ895tpvk1yYDu26EM9"
    "GOIGvoPmqb1LqpG9EYCePcd64jw0eDw+KhbVPQeCFz56fRo+FYPOuQgNCr1tiwY9hqzvvHxq3Loy50w8VLiHPb94sT1POCA+"
    "/6tKPHF+yb3Zeds92EaLvfkNIL4pSpq9cxL/vKCLl73nMTw+VgSqO0CsmD0oqRI+UcSTvo8ITrz6mOE8Q9wavZ/djr6AHB0+"
    "QMkBveZQML4ItJI9BjEMvtahIb66y5S90SXGPbQ4MD7glWi+nIMmPSxksz3zVne+t+JZPlV4UT6Y4hA9YCsYviMZszzr4Ro+"
    "kzOiPh97sz1imNK8xddNPvBkRD5uBmA9e/mZvQ2Kwb2i9Zo+hxO3vX3CAz72aSE+cAWIPRGWGD2WTog8OfQFPsRU+L1taBA+"
    "wo26vaR5Mz5PQTU9NqmXvYDdsLxiDcE8AoBivcXFzDyMAyE+gGudPcKjND52FaA+L6ylPZJ9Vz4/7n69vM8VPVNTC76j7q29"
    "Ts3jvbaC1r3beCO+orw+PciBzT0t8i8+QIJvPmGLZr1JPwo+Rq3Mva5Rnj3jXkS+rzHKPIqj7T2xG2U+8KmOPieSgz3MPZI+"
    "mGWSPj07gj7MGNo8+dIAPhoOhD4vQRk9wnEWPgZAWLypgCA+TZwkPe1n8j0NnD8+joVLPZy6Fb0FXay95Dz8Pf9EsT3sx1+9"
    "73YgPola1D1+sWS8FoYSPmJEPD71Ad88G3AVPsz0/T1BwbI87ze4vKaQ8bzYLqE+JyeIPRizWD5PCoo9NBmwPS2A4T1Mtaw9"
    "hZ3nvdKN5L3S7Vk9LNxovbRhBT5gBbk9eXgiPkzEJj315xm+ySixvaEBh7sPd4Q9OKZJPkMWVj6By7E+l6o6Pacrh7ynPHa9"
    "hiUUPo6JCjtbc4K8BfIAve9ELz0H+DQ9it0tvdIpkb1Q4AA9XfCLPXklGL50qjU+5eISPVBU7rzdR8a9plLCvcjrqb0XTlI+"
    "lR1PPoTTjD66s/49TERKPp07Mj47vy4+7r6bvQK5cz4ohkY8oDjlvUNSPD3FBJi92ZJyvfrgsb3oYdi9yr8pPlBLBwhALRM0"
    "ABkAAAAZAABQSwMEAAAICAAAAAAAAAAAAAAAAAAAAAAAAAsABwBiZXN0L2RhdGEvMkZCAwBaWlr9Uge+KELkPaX3Kr6+MVq8"
    "gG9DPOxbMD7zJD+9ivFlPXLOKT0GdSA+7HhjPBgryD1iHNK9qEGwvCmrPz6bf+a90b+iPAHZMT3QnFg+JxzevEbqG74QWiO9"
    "t2EiPhUsxz3gj00+V0Y8PXm9zj3lkwm+MkS3PTZ4Aj6CUSo+ccGHvcVsQ732mdk9ySeyvXntB77wiBy+oOs2PG+8BD7y9Yw9"
    "xH4JPRmEMD7nSNW9WJGjPZtLMbw24Ju9O9MsPkYm0L1muwM9w1gdvvOg/rvHkMK9t3WTPapNnr1tSxA+51C6vb8BHz6PO4u9"
    "bgjtPYnLgL30gYe551yuOx0uBD4QDaw9UEsHCH4D9z8AAQAAAAEAAFBLAwQAAAgIAAAAAAAAAAAAAAAAAAAAAAAACwAHAGJl"
    "c3QvZGF0YS8zRkIDAFpaWi1F8DyEpGw9wOBWu4OcFD4wPqg87sVAvcYOSTsrp8I9YEqgvIlLJD1Y8gQ+skAkPq6j6D1E1XW+"
    "3q8aPkbMbT3B9ra9m5HjvWs85j3jEf89MB6ePFOF/D2pxJi8tlu/PXDSm71lsIS9gvmwvD3k1T3VDg8+yigYPqKbHj1fAIA8"
    "5wljPQ6Tb7x3BqM9itmDPX+MIDx0j1o9EcoxPhkFCz2qF5Y9muo7vN6Hez0P+Pc9fjBqux6+Lj7jWbQ9nRtsPlRFOj0Y4qo9"
    "f5bkPSdCrr3kFZU9vw8RPbmUAD4/kJM9/eMhPVpWHLtx6y29aunevAaAljyuASk+Itg1Pukt3DtT+yE+SeHwvA7m5Tyjtio+"
    "+Ex3PXfyi70ki8I9NhOyPWNIwT1Xrc69XOUXu71BMT6E69Q91ejXve58oTwZvig+hWGpvHUqab18/E89JU4aPgBUj7lw4bA9"
    "BZ9dPlYmaz5p2oG9Vw0RvWBeAj5bDgg+aDEpPtS7fL1l10o+/MxZPpgmi73n8Ui+jSqivFXxxj3D53S85HorvTLTwz1UTEU+"
    "KEufPatuYL1DlbY7Kf1dPTMLHb6K2rY9jrCzPUI87T31h6I9zwBKvTFGFr3V/g++suEWPu/VBj30NX+9QVXkPdZ4IT00pgW+"
    "5n4HPnAAHD5YoKW8tjLYuwLszT1JnCS8LxDYPIT0Gj6N+o89jzXcu+ifL73QpQW+eqYuPu4k6zz5Hze89hYMvp5bDD5aYI09"
    "NpCSPfJYOL65Gky6yVUCPpksSbvAg02+Tu+yPHqXi71IE3C90v4sPjJE0j1jsGM+DDBdvD1WVL61wfy6UzWHPnYRNr1NoaE9"
    "l0IXPtsy/D1GP7s9IRURvh1LOD4P4g88J4ZKPRBnuz1HP7U9pR2lPZ7wnLwMKUO+KmYNPaNVCz7Q1j29sjPmPdwljz1C+y89"
    "blAtPfnX+j0RzLs9K0nFvSRoor3Zo7s9+w/kPeFEHD4kaPI9/t9QvtbUDz5peUE7eokjPvI5Oj4r6Aw+f9NMPvYoZD02FJO9"
    "HoQBvbDQDj7UWzM9jYEBvn9DDj4CnjQ+QXtgvcJPRT0gYbs9atDXPbeDUD41ZVG96h70PTBwUz5zvFy9ESb/vetKv72E+FY9"
    "WCG2vc7Xzz399fg9jh/kPXuE1b3iJ2O+he0uPoU5Xz6vzvk9KQsQvfIjrD3zMWw9mTfaPcNCKL5R2Dw+VNNGPtCUurwAGkg6"
    "kUYQPWmYGT0LenU9y8MKvg3xhT3BFH49FENAO5XuwbwH1sC9mn7uPQAxQj63GsI9XcacvTsSwbxUaLC8PjFiPT4Va70FmWw9"
    "OZryPWskab6EzwS9hcXUPT1+4j2HN409fcFfPjkE/DycaZu9mFq/veiWvrw4qgK9MmaePYD85btQgiM92MNUvU6qzz0U0Nc9"
    "AkTlvXSVA70gN9m96IJZPbSp371kFHK9xMn/PZZ19L3C4Lw9YAMtPNy0zD3S6oG96iWoPVAZdzxAhB08brquPcBV870OMs69"
    "Rjz3PSS6mb3ahri9wEXIu3jkI71oGIy8LISrvUC9OD2GSNy9oN8lPNhilT1Qv8Q8CA+IPEDGlTyc/RW9+BHDvDDoV71w76y9"
    "cIqHvU4S/T2AX6u8AFozPbAJsT1A/B+8rLiJvTJzrb1cnzy9YMA2PbSX173gDXq8CMJCPeLb0D2QExo9WIPKPFZBrz3Ilmw9"
    "OSE7Pj49FDsKy5+8f2XjPHC3ubwY8Ku8hVqDPe5f3T1QHBA+v0/GvQ/fT70jg8k8wWHlPBrBjL1M9au8E0C4PQxTCb5m60q+"
    "Km2fPTfCg7wg5dO7WnUOven7rj29uPU9kE/BuwNTRb6eOeg915z8PS88NrwwQJ+9MZW5PSf487waXfY9vn0qvscJZD3xg+g9"
    "m18CPmAcQbxJpww+q//QPB9SfT2cRvS9b/P6vdEPwD1bYpO8dhDbPX4xcj3kH0Q+WgCMPFfUlbxkqQK6jKoFvuthKjxF+xK9"
    "t77BPdI+UD5mJzo91kFBvsI3e7xQpA89tcAEPk1Buj0owPc9vgRfPpJNRz69J3s9z11qPZLU6D10Tzs9+r4YvkkS/DrOdXo9"
    "9k7cPSqJTjyEav67sXsMvaCSoT1cujS+bS7IPRMPjT2U5oO8xCkGvt161z3VrXS9hNFMPdwBjj26hAI9JPspPrtbBb7x8EK9"
    "v3KvPW27Kj69aOI9LnhbPEolLT4MvSg+E4SlveGkFL0uMqs9NEaiPYD58j0YO8g9kLN6vHzojTwZ8q09ytJBvi0I1jpg2PU9"
    "2DW2vOC1Nz4GpMW9lYA2PbgWFT69mys9IEn+O2p1Gr7vOxk+SuJvPajvtTxrClc9AQC7O3uWZL3kEaE9RUNAPvXRrT2OoZE9"
    "dxUwPnMUyz2JufE9JfuxPfKxPD6Hhqo96EGyPSXagT3KJ9c9ONvGPX/rxj27GS68Am0vPrQOTj4jngQ+9TEQvjLaMj5go4Q7"
    "pf3uPIrzHr2zoss9AhSaPEYAv720ER8+aiZGPXthBT7ObMs9mBZvviomoD2Rq949BNwRPiOLbb37n0Q+TrQoPnqI2T2Rikg9"
    "+w36PS6tjrscTDo+4Hw7PM8/Xz3DdbC8eDSRvBpWEb7NPqA9GltoPmk0nD1U1yg+FQWbvbViYz0uLRI97BaIvP6G9z0rwAQ9"
    "+CUaPtVaYj04pIq9CyK8PQSG8T2b7FK60ujoPY/jMz0zcQC914uXPEiYJT7BLDw9i4hQPkfjsT0mBCE91M0LPrSFJL1lM/68"
    "doQVPcSMZL1Sg6c9awAEvEawRzwEPi0+9b2GvKMCgb7Z0ne8UL6HPVhu6b0/kwq+GVuvPbYejr0QyJY81A2HPXqNDj5lkGw+"
    "yDPFPUOmVL56Cri8yUYGPjuJ0joPU5I9SQ76PVSjQz7CcpE95U8Dvksz9z0k2RQ+LfqjPUSTzj3XMS+9Y6MlPkwDZr0GAVy+"
    "sq0/vZA7VT4h5UY9Q8DsPQnpJr2Hwcs9ErUyPT9s4jyuP7k8heuLPdto+T2QRgO9mjBPPb0zQD3hvag8PLdCvouI6D3YpAW8"
    "IftiPq2BPD6/6Yk88jt1vCw4BT6t5LI8LsjaPUnQAb0slVE9aPASvknqDLtEm4e8xt/bPZD1H72iC4W9alWfui18oTxHiW6+"
    "GZCKPYE8BD5fTe69Dvozvu0ryTxo35u9+OCbvS7ZZT1JlVs+1nlFPaX3wT1GJjG+u9j8PNGjFr2EiCQ+/HkxPqRR+j2qZRQ+"
    "Ft4DPkSxRr1dtys+PQjfvA9K9LyGfcS97digPaMOG71xroS97mQxvrQsGL5fJvc92zGCvMbfmDuGzns9IgnLPAKqKD0J2Bs+"
    "iykGPjpEE73oSl29lpHKPNlQiT0HrEk+E9XCPfOMGr4yBHu9VLVSvXzoRD2LEpQ9fJLQPEu6sLxaVRo+YQSJvb05i7z93FK9"
    "CL0KvQ6olD24e/U9r5MWPYFxpT06fCo8vnt4PVi+f7wwNYc79Kx8vtxuZj1Po3m8TkkDviD8Pb5fdQg+naIBPtSylD1NBnU9"
    "URdcPYB3BD72lcC9bElLvmCJOz3Oj/09IVGOvT6TYz02V18+cnSKPMd4VD1/JwC+l+3APYke3z1qbU8+0DSJPNVZ9Lwtx1Y9"
    "zrmJPSofrr0w5Cm9dYr8PXH/yr0YkBE99kW9vad8BD4H1+s95N0CPr2DM7wKYMS8GS/6PeDd8D0wV3M9iFAgPgOJBD19NV++"
    "geekPZRYWD0nuUw+XOJLPq8k+j1f4lU+SNiDvSqC3j2Gadw9oI2cu0Jxzb10OCW9POwIvcBQBztMGui9+MW9PMDvVL3UFwO9"
    "1jiEvYAz2LzuwO29FrmHvQCsPDwqwKc9sqPoPRKnwb3o/1Y9Xk6QvZ7txb1gXrS7gtqSvaR63L3IPe+9alqrvUZd0j30xY49"
    "eKxKvZpYq70I7cS9aiyXveivUr04wJk9KI4nPYBPRb0KPbA9JAmfvWSP0L34fN88nCiNvSinET1y5Ng9qCByvc6jsr2Et0A9"
    "pjTnvcJl5D2Qy3G9DIh3PWCQy708R629wJhhPKD3NT1MgMY9HBrWPUCR67sgp4O84ESpPLDd7bxgbos7wnHyvaFiJj0qbgk+"
    "cDZFPH44Gj5QwWA8s/J5PAzSOz5MpIy8k/oTPZU7Jzte7tk9jjgavdbzAz4YzlG+lDOgPXa9PT6EwIK9wvHPvQcEPDzHlwY+"
    "Cn6RPZRStj28pfU8XfXiPN4LDT6hyZW81dUPPvPJXz0rkxo+ijeQPNHxVT5J/kU+icpgPUk27b0SiGI+AOw0Pb5aMj7SefW9"
    "gu07PF8KWrydsDC91nWTvAI1Cb7difI8InynPcJ5FT5WMcQ7qq6rPAsJTz1UWdO7MJnPPexjML3miV88QfLbvBYOe7wTIUo+"
    "LlThPaUOdL035tI8m79LPiLf3T1UuRM9oKVRPQBqmz1uDVM+NbzDPUWNNj7hpD49WrzTPRYIjb1lLnY8WWHUO+2se71JqsO9"
    "+6VFvcdCoD2PcrU92gIrvkcUrLxVu10+IXzAvTgwMb0v+tw99hRdvRAGgLw/nxI+2EZpPk35pj0BsgY+R8tWvl2IPbw1/YM+"
    "wjwyPeoKIT7QEwU+A0QNPYr8Jj4xay49qlbxPSGGFj1uOL09vDp2vfoPEj6R0dM9/dEmPUWnOjtlaLC77DyDPe/Bxj0DyyI+"
    "skeLvWowVT6xc/k96rNlPByPh73/Xp68riWSvdlpDz4mngo+V/3GPVv+FD4rXwC+tw3APVcsfLuf/KY9Bm0LPlAnGj49Vbo9"
    "GaGbPcn5/j0jIyk+2k6SPdx0a70kfy27jcOQPZhoKz5wSJk9ck7SvMEckz2Ynhs+PXWmPZyHIb1zOQA+OI3XO1pUvL0OI2++"
    "8gWpvc5HTj2oHOY8XFFDPWuLYD61XQE9Ub9LvfSCH75e45E9KysRPiNCJz6oiTa9sdJHPd6+xT3VGO89LqCLve0OLD5GaAM+"
    "+iBMPnjCmT1PUw8+jcijPUn3yL2m8y2+40Mou7kLZz6t1bW9k/U8PjmvIjvpPd495gizPTAVQzw+jQw+GxfEvNJK+z3Rfii9"
    "sZWzvHLZBz4wCgM+u5H6vGvuyz0sDjK9ydS1PRPNYjs01yS8zwrqPexAIL5gfyA+lSanvU21hj0yQ5Y9MlbKu9hLzj0H0OI7"
    "yVmPPdUzIj6ldCm8dfxAvZNPCD5imRQ+gZk9vITPXT20wx89vEPdPbJ0Hz6hIvs9qJWlvUdDVz1NYpS97gjmvTc86j3tM+o9"
    "HoWFvWQDib3m3LA9s2SxvCPL0r0n1f29AQIlPZxtGD1yGGQ8pcQzvcli473sYCO9ncKOvWB0wr1XWZw9ekTNvIMVmjo9uQY9"
    "bLmRPeTLhr32+Vo9aUZiPWay5r2CjAi9zjRXvGNMbz0AQwg9FnvKvbO7/z32RTy+7rlCPpLSl7utgke937BZPYGghL3s2V49"
    "XPIFvogrYr2is0s9M86mPRn9Qz68My4+MCBzPLBkrL211kQ+6KOCPQHzET6zOLq9aaXDO8LYmrzgE0w+PPbtveWo57xv6ek9"
    "MTr3vaBE1b0n93g9QQUSvfivsjxtQVO8puMCPkaRt7yWYUU902HoO9RxPT6RIEo+EY8TPnanFT6nfhs+hgACPvy5Uz1XKPM8"
    "hsNXPpCjTT0V6bM93vuAvaMYxj3spcy8W+tkPWB09L0Wsc85nnk1Pofrob1BABY+BX1XPYQDCzvqelc+IuMTPqduzj0Hj9S8"
    "a+C9PbK43brFcxg7PFc/PpKIrj08Jwu9L8I2vZ3DOj58kgk93q1HPlKgSz7nexE+FHR+PqRs5z0lkiE+Ck0tPLBv7T2hEry5"
    "yTUePmwQBD29QxU+AHkzPQ53ubzq4UU+MVwjPAIcXL4l1C296/8OPrZoVb5LZSW+WO8kvSbQjT2yVPu99z3Bu0R3zT1T3+09"
    "btNtvfwa1r2M6zU6ci07PaefOD0QgnG9OcFEPreawDsocLg7XO9fvReptDusVE0+r3kUPdRSC70aAMG8LW9hPaR7A716Cmi+"
    "+a63O4H4DD6NipI9GNknPsmkkj34p/c9ix0hPnJBGT56dAM+6YkrvgbxgDxMg2e9pSNcPc2Mgj3hiBY+M/6YvcaIqj3KZzG6"
    "u73ZOwTd2T3bwiM9ZC9APYPzZT5W/wW931O6PMf7FT6sCUg90Jrbvfz03D2UC+w9ZPFivSvRi7w1Irc9cWmkPRBpTT5GrHW+"
    "X8EBProLCD1gaVm+CS5XvW98Sr1xW/s8iqGGPaFDUL2IFDO8d6AQPkhF0j0TjyK+RdEiPm+xSD7tuaw9doqbvWT7bj5FjBI+"
    "TEAWPX6TubwBXx+8k2sDPpWfDD0w+E+8/3i5PZzBGz74lAg9gMhhvmIb/L1goA8+rhyTvbogybzogMo9/yH7PUdFCLygVLQ9"
    "9gKTPQRxxb34y9Y9tkRZvE+5sD0RVXM+fz/fPauZMb2RNVE9YgjUPczIBT6hm+898F+TPfdN7D1uBQS+/SHAvDKbB76lLoG9"
    "8rGgvQjdUj6ynpE94LKsvRW5Lb3vjIA9RZVGOgHGcL2wwJy8YbTUPfIUZDw1eHe8qDkkPpC2Qz5ExDA+qwWDPTyeZL3Tl7s8"
    "ZY61vWG1QTx+s4k77rTRPcX4zL1jXfe715v0PUsWLz4ElLC93gcSPNzR4jtCvns876yPvbC7BT1mKNy9PGq6PX9bjj3DTae6"
    "DVwlPllWCT6f+yg+8HatuwQWdLmjQKK9V4SXPOAD1jzRNac9FJL2PS9mFz1hDbs8JOBdvfu6dj3UbDW9Ir4Kvn70LD4wqOc9"
    "BUOsvVWvtL26K7u9dVRnPT1RIr5NDVg9/02APoY8ST2svJo90GkgPrTHYT1vQrw9oNs2PpR/k7wppRQ+QQfiPHIlTLySwh0+"
    "8mLlPLeZar7aT6E9p0u/PTL1EL5zzmK+6/ywPfKXHT5Y5uK8+oUvvIQxAD4zAOE6q3PKvXIZPr5xpiM+CoJiPUYHKT3XbFm9"
    "MFg1PMWOiz0va6g8584DvjbaSD5CujU+VaBAPnyLSD0qsZs9/Es1Ps0qhb1I0m+93jGtPLk9pj2JKkS9bJ3+O71L7r2X7fg9"
    "lbrlPSL8AT0FiIw9H5Xmva/7J7wlDRg+tjORvXzL2j1YlAU+IQlovKuRBD773nq8JaTpPbm9PT32FJg9fIFRPmvjyT2afz+9"
    "Nc6QvE1sCz6ApMU7j1rBPRXA2z1zaGW905i5PSVPjz2Z7xc+kkICPoxDY7v8XVi+x/AUPp03oTyiZqG9JUcFvluUhj2DUT49"
    "1PzCvVdjeTtXrVc++LrzPRWGgj1q0FK+VnM2PnYrBz7AKJs8F6gqPbSakj2E6h89iX8ju8Bc5b3GzyI9pzsVPo6a17wkJp09"
    "HXktPnKX9j0fthy80e1tvkJQMz1SPHs+xpVqPSdvvT1q0eo7V+1pPjAz2z1i7u685Qe4vek2KzxEqL89YocxPjh9dbyFjSc+"
    "1Fhau9ecDb5nIh8+LRb6vKKFQD6dUxY+SKTQPBe1KT5o4OA9ti1cPWOnZD3IQZG9cBtUvPkXBD1ho2c9wjIavH8/dj1Br++7"
    "nejnPIpwK7x4zY882IsqvuZc1T27tVs+GP+evTB4m71UfrI9BJPzvOjjKb1MJhk+lxVGPj+AAT76/Km9GV1qvVQmEL0tYGA+"
    "H1iGPRznWj2fzxq8mtUdPShMnz28rBQ847+IPbwBaD6IC3s9zmeavSjmn7vKBts9mPyNvNoZEb1/WgC+q/bQPVBGH74+Xjg+"
    "ChQfvQSBAjzUKKE9NHz5PJzrXz1FIBO9W+wAPq9Cbjw3HiY72DrFPasMQT2CGmw8da+KPRM4Mz7A6jo+fD6PPJkitT0TxAs9"
    "hcVAPiY3DLudmwM+I+6YPR4OqL2VE+29h9FtvRtdBDydgeU97gVOveInOj5Vdya8t9TAPSULgL41n5o9L6caPrVsUb6Gmqa8"
    "rjOlvOK8Ej4o4au8X6ENPgm0cT4ghJ49mJt9vENqxr1Xr4e8nQ5hPsy0Ybzbruc95hQDPah2PT2SP/Y9sO0gvm2pYD7uj18+"
    "8JRVPt4i+L2R7JU9HWuKPb4FBD3Eh06+J6vvO/DzCD4FXee9/fqvvHf5Hj0/TJA9AE1NPudegD3M5pE9i82JvU9qNT0eh+a4"
    "Ai5jvfaLOz6h6Ra9ygQavNCPYT01ECc+CztbPi5pGzzitWA+gZr9PQttST3d/cg9OxgjPkGOJj4MsfO9+TG5PS9uKj4JQDU+"
    "sN//PYEkvr2ryiM+37hNPlnyID0vQSO+8M4jPi85Uz6Ed2g8DQGJvFIWOr1GJbA7bIxMvbhECT4fIw0+cqMJOwCkOjyuBWS+"
    "aUtDvR9IiD4VHCK86g2ePZMntz0UUno963wPPde+kjz69rQ9La5cPvuBKj5Enq89izl+vZEbKD0Dv7y9+ozkvd8pnr1fzCA+"
    "M2y+PcTVCz6Jb7m9owP2PPD1OT7ejkQ937tqPTSWqb220w49w1XKPcMEIL1GAVQ+STC6PXqJIb4Y/L082rRRPS702D2O4SI9"
    "hShePtN2wLzOZu69k/BqvdeXErsjS0c9xOURvfyGKD4h26m9eO6cPeHjI729W6k8mYi3vQLCq73X27G9H9AtPltlWrwEgG89"
    "vawLvYdrDj135I09xWkgPrD9Cb3wmoc9wnMvvFeXlT0201g9bEIZPtJ2pDp1tI68aKrIPVMF0D10AqO8sraKPQKT4L0Xsgg+"
    "O3aSu4UmCr6KhE29uoHDvVCegT3zmI+9CdEVPVKRfTxjEUU8AfjKvGnqQ7zWBW48MeuxPcPCo707r3s9mf4kvK3hr7tXZMm8"
    "qK5QPR9rALxYgzY9k8Unvnlo9T2ISqg9JdXXvbHqery4Lq+85BqEveTjn7347s29/eODPppekr11X08+qssZvbhw+LxaIq69"
    "yhIQPMJ+4j06Ig4+HV0JvifJnzxJMyA+mO2lPQXj8b3M6AQ+wtCEPVBhPL4bgJW9eiTPvCIKQ71iIYG9kTiiO3HCZT7pcAs+"
    "zdthPb/TF76ui/E69hJ7PlJ4MD0esuO8tXJgPqgY1Tw6Qw09RR1nvi4aNT4TMyY+X3wKvBLijj0hqJM93mUgPWAM6z16FIS9"
    "yuVOO7kQzj2MWgG+E22dPO2o8Tvt8PQ9Yrm4PEKdGj51mc0934jQvXMK6T1xql84VZyxPVzDYD4N2Jq9B5UlvjoupT0T/9A9"
    "gfAmOjgemTvAH8M9SqW3Pd4XLD7VGGc9ulPmPZt8jz0wZGs843wYvdYj+D1lLxY+L+H2O/FFxr1jzgA+v1vyOseCLz4yJ929"
    "d1+6PdunbD1J3Lm97AF6viGmyr3AdxA+9pzlvZmk/bxfFGk9bEbtPWIH+TyCLk2+4V+GPaDkAj47YiM+Ku7rPed1MD4rWLQ9"
    "y1/yPZPSUr5Y6h4+ytY2PWc2SD4GTeK9LlTbPa9eiTtqGaE9h63Nvbz6dT1nBkE9Ft0HPbdASDyQIaK9ALgkPipwhz2FyTe9"
    "HvajvRg/VL2FJrk9TIsvPtgpDz4eqNk9r209vamAZ77Z+EY+TpVxPOuTkT0aqA68fLOvPJksAz699dE9jgQ7Pfi1nT0GCmw9"
    "aObKvCklNb2HAf89R88SPrstJj41FBi+io05vTtKiD0Z2pA9NTG4vd7TMz7LJwY+oHdMvny4H771kbc9tmqLPfQLtb1113Y9"
    "DT7gvOuORz2quKI9HtZOvibkFT4FLEs9PKkhPuJufD3pskc+YEH1PWMDzDzMcDa+xmbkPHwuczzgBEk+bLQxvVmZVj1cwqa8"
    "ws0XPUN2Mr6C7RW+x8MdPmgCLr5I1JA9C+MFvnUCPT5H6Ka8Fw9gPVeDoj22CQ6+ObT5Pd/QKT51NQq9IHP6PQPrgj1Kxe67"
    "JwMyPpJi0bzpfYG82Y7LvCPSKT0oEZQ9B03dPS+6Dz55XqI7CQ3vvJQeyL0KrKW9CB9SPhEKsz2w12E9UACIvRBjZbwmmzI9"
    "YeMoPhlCc70y3Cw+SEgBPnAGG77FRUm+kdgYPpPEmr0obfI8L9ahPZ/rfT7MmkE+u9sBOyFzb75LNM49m351PTE+Vz1T22I9"
    "ToY+PvdZUT31LDa95vZdvjpX8rtT1xo+0x3ePACMlDvrA/077N1RPOd6zj3gLdy85+Y7Pa02gz7D5O69iwmwPCxTwL00Ak0+"
    "g/wwPigfLD6jk029Lr29ved7Qj7ROqu8y7TzvKZUCj5SHo29Nvf4vZ0wET5kRUM+NxiyPefHtbyPxzc+rstuPvCGAT1SBtE8"
    "gIRSvAPEJLwwTSw9uHRIPGHQ9T2e6m29URv5PFqIq7si8249HJxRPQ6umDxuDuC8J4MBvRGb3btwnZm8AsNavpHmrj3F4D+9"
    "OIaKvKBxJ7xsOUY+NYtGPo+YFLwW8CK+Hb8SPU4Uoz2ylUc8DO0vPds0uj2TWzE+X0/nPL0W3L183/48eWoXPkwxPryoP+88"
    "tNTMPW1uAz4wc8U9tOPUvabEv70RQFA97Jw3vuowDz5qhtQ8vyw/PIhK9z2V88m8EyEDPhAPur2qFxY+LJ0BPpWo7j0Qqyw+"
    "z3iHPUQgjr3PR7S8Md9QPTLtsz0DcxM+tFZuPYXlJT6tKWw+0pTJPb7SFz4yjTo99EJ1PQ07ib0Bwz8+w3R9PWI3fD0cCA++"
    "iaNCPsBZlz2YVj0+eAtpvtVR+byha7o9coc/vow81b0kUrs82rhTvD5KoD2p2Rg+ExMKPGj9KzykGKQ94r5avuk+2j2XVvg9"
    "0oStPbVvU73vq/c9bY1CPVn0AL2riSC+FrsLPoUXSj4KzNA9gAOguvWGKD7uBFI+gwZYupRRYL580E471sU5PvfjEb6eni68"
    "dV6tOi9Kcz4aYK28V1Xnu9QqOrycLAO+jw6wvMV9KD4Btxc+rZBLPDO2Ij0SKVm+XVsMPvk6DjyZphE9pKYRPWGN/z22xEE+"
    "2HqzPUN7cz04Ijg+KVtAPXB4Uzwfl8e8bYyXPeOTCD4DcAI+XhLrvd+OPj7M3CM+ygOHPTUgHL1PIow9Gp46Pia5Xb6UII+8"
    "76IJPv25fb3IIeW9fGCOPJJptTz6eQw+LHxMPWNjJ74jbCg+nS0lPsqalTypwMy90M8ePjXG8rytJn48AiC8vSbVDj2l8xI+"
    "yHH0PSpSqz0QLAY+dXmiuVzmpL246TG+XL7wvbp5NT5bGgq+kMwTvNgWnTuR1To+RABKPVICtr04mic96Yi7PDWTET7CcJs9"
    "sdwIPri+BbzIhhs+J+E+vmmNcbsSTTA+Zm6GPc/iKT78QVQ8SUxJPmyPhj4Kx649FidAPi1f77zIHOK8a53zO0JeHj41ElW7"
    "emERPvlDxL1yN/w7BdvrPSK6grtrmn6+yeXRPeiMGz5rsny932z9vTYdiL2hV4C9UPf0PaBiQr2IhBM+1u5VPuZW0j0sI2m+"
    "WwIsPfxLaj3JYBA+iCrMvYT4HzsVcYk7vU7pPeV3+byN1Ew+IHxUPgSrCj2YxI69K3m5vAiREj7Acoc9AJKyvUYG9b2bsSA+"
    "IMHqvY1aZD3jZ3692rDdPQMWfrxp2Ag+uK/CvH1uWrwTuAY+BYvWPTXZCT5FMj4+56EBPo4ElL3ZKyk+isjAPA5RAj1T2hc+"
    "TKkKPu0ryj3DFBU9VrWZPbU5LT6+zR89KEHrPNtSoTyZAEu9NbUMPQijaDxZg+W9OSkRPsSDXT6edt09JxyjvSyjybwCZv89"
    "IDP5vc+0Xb4AWyI9ZNuNPSDWxbvZIec9J6UaPnxTXD5sBHc50QMQvizclD3zdi8+elvcvNJkOrzfCgQ85X9OOxA8pb2OhEy8"
    "oMAIPfas0j2zfOI9fBriPWEq5D03qZ48Rv8EPpKPeb5xtMK80HOAPgwfXj1NoWg9DCPmvYB/Kz6rwjg+3m34vDPuCDsDM969"
    "P7PFPTkPrz0ez9092yuMPajltb0/vh6+q0iPvHXYTz33tQo+eI6EvP7Byj0tmKK8r6Y+PXhoar18Exo9qKRQvFCenTxeGhk9"
    "zQYpPuOyuz0AzWM9G2OFvZ1wFT70yHY8B4UvPp1uj757hAy9NroGPiRFJL50w++97NDyPc8zpz2akPo9iP3pulNG/DwCBqo9"
    "xTAcvSJtTb4Ea4I9g3WIPrLKorwgkZI9h2YuPm5FIz45/UY9BsMjvl+lGD5dqSc96IeIPLpO7z3tWh0+viUTPLUc2T30bNS9"
    "vMBMvFLbhD5WDX69SjErPe7sST2BTpo9UTdKPaK2pLzXXBA+yum9PBD5rDwEznM9zFBTPf3nBT2Wpn49z0tRvlMVkj0ynmk9"
    "7iEUPbQVTz6HPwg+ahAiPi9FZj3O23g9tnbsPVXi8z1oFBe9fEhdvSD/+bxCAgY+jWwRPsuGuDygoAE+/r4jPkzETT7KNNi9"
    "K+vWPZopEz7IQCW+Z7XAvQq/WD1Sipi9iIcXvTQagbw66q08nk8/PoiK2z3ERxm+PoHfPTC7+j2C75w9X0ENvNUDSz1KIQM9"
    "+SkXPvOzpb3jEgU937X4PRuQBb2AhpA6XIsWPuqD4j0wVxc9AeFlvsdpjr3KcVg+B5ICviLmuzyzPo49PsQRPolNEj5ApaK9"
    "RkHNPdhMaL1Q35M9JaQrPkX/kb3UdiI+4VSuvYgRqrpeHOc9rh++PVZPFj1u8Qk+F/oAPdTW9z3+vnk+SUJtvbtVQz19P609"
    "iL6xPI1NEr2oFzo+tTC8PEHewD3vEdE83OYoPU0ULj48diw+dJPRvdCwCD3ot1o+lMYwvgZV0L10GhM9+JrhPZDV0b33Aqs9"
    "d2zfPGH8LT7sajo9GihzvhLhhT0skkQ+QMHUPVge8z39EDM+KpvkPei7tD3QG2i+jXI+PrUMwz3Qkyk+LJ4SPUjNQT0W/UI+"
    "kU+LvdmFs72GwoU7UrvIPTmde7sjoiK9yiG1vee6Lj5D5Rc+/1Siva+Zgj2/xZ28voQrPjPjib3GQG08jXYvPcghSb2X2gS+"
    "/KESvYNUNz6YADg9GHQGPvWA7Tz3XNA9BP9bPajfFj5W2UQ+gOQaPuw8M72z/189q4wCvdEB5T0FeCg9C1IKvlc9Dr2Nefq8"
    "M3HhPbDIOr7xoIE7+MRlPu8j7L2NaYy9c/64uhkn1z3wyRg9BZ4VPo8UCz4vOpE97mQKPgC/B75+aYO7hDJCPt3cHj142JG9"
    "gQ4JPq+IaD2925K8bfgVvrmTVj3CloI9KFDYPT5Q9z0yzbS8jm4fPUmrkr1IX8C9vSsUvVg7Wj2EbQ09u9P9vAPw3D28/AY+"
    "ZpCYPZtLzrs4sKE9Vt2SPe18Bb363Q09SO4KPu863D2jUXy9Fpb7vcebnz1x4HE9u4MDO9ZjADwzJAY+Khc6Pv4cHT1vi9i8"
    "256+PIjPrr36aYg9168lPoKEoT3XfAC98I3nPe4U/L3OtrE7TemeuxTiAb5sFOi4ew66veMYAby+PZ28Rh37Pda7TTxTZSI+"
    "IPlnPH6Odj1QJtc8ilL/vduTC74N4xk+5D4QPngK8b2itt48iahXvXtqYD0Zbgo+VzZRPVhG+j1L0yO9P247PAWvpb0Acco7"
    "oC//urfTmr0iZJu70BT7PIKA3D2zy8S9wBfIvNtSYbwLcM+9+QBCPYvAAb6mkRa9lJFUvdcFBr4Unpq8oQShPRxOGT5q17G9"
    "OYcSPleLtrzsHMI9o93HPYkV/b13X8q9uNdFPbjtUTzgemM9frd7vfAYLj58Ros9AAqRuW8ZHr4K+js+sGLXvAds6z2ZiVy9"
    "rr2NPd4jGD40bFM9b1FAviIKJT4oVAc8uZs6vWoEKb5g7OS8W9TiPXjgrDxrn3K95/paPiYrKD25Bt49+Oc5vqkvST4pqEQ+"
    "AfQ+Pa33ZT0ERec9VnrQPNAZsz2UMta8KNSxPW4eZD6q1NE9AmOHvUZ2gbwdhjs+PEEuvR32I71qkSW+OL0UPhL3CL6VOgE+"
    "gf2ePau0IT4PEy09FNCpvc9tHL1Qddy9sfUnPj3jhbuzpL89CTVhPlbzlz37nl2+oYSHPdjKNz2HnC8+ZcydPdPqvz2TS1k+"
    "87cOPmBpyz326oY8K/ADPlg7kzynikG9GzCTPXKGZ72K51g93AwyPaFskr21P4U63vMMPpOzBr52u1U92nGpPY9xwb1EUL87"
    "9MyZO6W2/bx6wYY96W1dvR0VOT0ALL09ger5u0KVzrxRd1S9odynPZKrz7tZICA+NSMcPsNgr7zBBa09TiCKvR/SQz6VehM+"
    "55uPPCAy5D2/M7c8mChIPTIgbj3vmz090IMKPHLoUbrgrzS9cDtEPn4WAb0cVQQ9Ru0kvbP+Fr2XaDW92IVYPT76qj1VnC8+"
    "VXwFPv1HzT19NsI90ei0vEjCSL1lYvk9yM51vKP4LT5NfPE8T+nKPT+lCT7YEsQ9JfTEPbWPJ70MaE89AGYSPVUghDtvJ/68"
    "24cIvaczzL2dpSo9oei/PRjinz0kyjy+GxwJPqf1zz1PrjS+qVgqvpQFcb3Al428anzxvcjxlz1AfKy6cXgRPvWltz2S9HS+"
    "oh/FOo+KUz1526o9V4gxPY1sWT0E8CI+DWWWPRQ8Xr7Ewrw9DehJPf7elD3Cio49W4sGPdNO8zs4INQ9QuNzvo9Imr0qtkI+"
    "+rKEPU2jTz5CTFy9i3kWPoY8AL3vsHM8Z3qlvOTeGDyEDy+9LKrGPZfH4z1T1F8+BD/4O4hWDDsrPEW9+9E0PcdlCD4FWO09"
    "xF8gPk6DEzykoJ09JkPwvZyIr73gRwu8rrT4vdxV871wQuO9NtLNPZie4DxQXSO9QFeou2Yi372Ex/G9cIeyvQyq9732bKg9"
    "KCozvSA2Cr0KdNy9kJipPJCimbwaSK69GiOkPe7I1z22eLa9aAuRPSBGpryArpq6nO5ZPYCxurw6gtk9EIKdPYItpr3Aypu9"
    "EqX9vczsIz1Qj8C95pLWvWwWpL2YGf29PHLQvdBX5D10/449mtLiPYxceL1g/1G99KQzvYD+Hzt6v7e92B6WPPIRpL0I0KE8"
    "0H7qPeDzrDzavLo9VlO2vcCCkLwSE569IDhjvAisS72UWpG94kf3vZDYPb3oIYk8gCS0vKJs7r06TcA9FtK9PbxXmr2gMFA8"
    "fvypPSYc872eM+K9qAuOPYBAIrtQCiM98OpUvXipuTwmyqk94IT6vPhgJb2qwvO9AAOevOg3oj0I8PG9gDMevHibuj3gIRK9"
    "9Co9vXhvGr3AcE+9EB51PFJ69r18p0M9AE8BvchfgrxALQq8rAKPPaQEn72y9Oi9ai25vZbMu73URnm9WjHKPaQLxr3ke5a9"
    "MLHQvdyagL3A+WW7mhmcvQAurTpUBi09yDiuvFDoXD2Q+eo8Skz3PaAd0bvw4909OJ2/PcjCdD0G4ZA9lHGrvXR0q71spVo9"
    "bOg5PQYG5z1YU4W8qD28vCghXzzVyh49SPGsPcHLGb38/Ce98DqbPUHg1D0pRLS9Nqe/vFsjDT4Z+iU9QmiSPZvNAb1zl0E9"
    "kmqZvVyuub3HlQ69RcctPtKCxT0f8K68FqbqPQkvKT0qYZI7gXLBvTh9xT1VvhQ98G3HPaqT0b1+N7A9S3gkucq7QT1aEPu9"
    "ICbtvD+UWz3ahae9ENW4vWgcQb3wWlQ8CCTePbtZb73ISEo+KHPsO3VuND23E4m9mBHKPEPeSztfBRM+wxgnPGq78L2Vldi8"
    "oLwKPfLhib0Fo3o9fby5PclE2D2Hk429VGc5PUwzAD21fBC919W7PZ7Flb15c4u8bsMFvoRPn73Vqxs+eKNMvftI9j0smVy9"
    "cCdRvUW5hj1NgEE+LZA5PWQJ/LwUoAS+03wSPa94xDxyhD8+KeVqvh0DDz7y2gE+8g3bvaN5DrwWIbQ91OzMPNyCrb2xjus8"
    "i0ESPpclVT6r8Z681LYzvge5S70ybG4+Q1o1PoOc2T190EM9QxQ2Pg9dTrvNaDa8D/U2PmFE7zskfwA+8GZMPU/isLz8hF0+"
    "aJ/cPXVhmzxMAEG9Xz+WPZ2ju712lfk9Tc70vGFWWT4ZSEQ+GjohPSj1oL1Mvg89y+clPtkaxDy4yK28ql7FPYGCQj2FiTK+"
    "UMYTPhwdCz4UQiQ9EOwkPakSKT0Lb0k+4WNfPjY0oD1soZ49mxyuPST99j3ZHSg837QuPEyi0z1yWA080kARvVMvh7zvpKc9"
    "GIe8PYjjZr4S0RE+6qlkPs36Qr5PW7e9rSORvVqY5DmgQXU99xKBPepeFT6q6Hc92ecKPsbdQ74cbcI9npIrPmiXCz5H3ek8"
    "yl5HPv7E8D2KV9w9MArFvcl6FT6003I+fZ9YPexYUL3oSxE+AewUPjjJ4L2Mimu933gvvkfslD2xiYo8DCkXPtcKK70Z2Vw+"
    "Ix9YPhR0Cz7pvMK9eSeMvRnxYD0DbCs+Oim3PDrkHj7/dr49lcEmvs5/xDkmw1E+XdmuPTT7Mz62QMa8Av1PPdRnaj43dg09"
    "4UkCPjsZX73wK/k98kGlvbkYdz3yyfu8eVvPur0QAr6kuEG9sO9uPXU93DwZ6Cu9CI11PaSdST7isZ29XUuevcgVDz55V4+9"
    "FM1dvaoUMj5sr5k9HMV0PUa/Oj2e3Ti+nROaPQAsxz3k/7o9YtbqOfpLmD1l7xg90j6TPdfVlbvTS1O8lJ3HPAwE5Lzo/249"
    "ivcSPoBVzLy+WTk92otTvW026r2uTCg+6y2xPRaETj65dHU9nSGoPTOyRTyOUTi8vKW8PM1dULziJg8+YmROva7dXj10vLs9"
    "i04iPeBdJT2FVZc9PI9sPfoGJD231Su9/VKUvAMbFj7hJoU9ecPfPXUgNT24Jcg9+uC6PW9aYrzlj+k9giAjPsfLlDzIFXE9"
    "M3MxvetiTD6E3is9TcIjvpiuBj4Jwuc9fMuWPDYRP74rzGS9FpoLPvqEmr3GdRU+q5JjPqNnPD5QKim9nhINvgHMKr0rla09"
    "PxHFPBnpj7188EA+Gq0aPnVkWbzapPe9BtBiPsYzzT2o2Uw+1CLtPc0c6Dtd3Le8ep3+PS/qNb5sVZY9xgBzPWK/e7wQxvI9"
    "iy08vc3vUz56BPe8UZB/vK9SCLplG5Q93gTSPeFEFT4i0sg9jctyPs2UlLsK0ha+3ZlvPU2fBD5l9049EkQIPpuxBj5dugE+"
    "ATF8PnXQjT2UQiw+viaVPKQM5r3ZAno9LjL6PaukIj74LNY9JciGvPgzz7wcgwq9ttLhPapD6Lzz3lg9f7p+PWc7IjxluiS+"
    "BJaEPYWIl71WAei9uHt4u+CTsT25eg8+CTbTPWunQr4VbzM9pHJGPmO32Dw3pwI9fCpMPiIxSz33t9s9LkASPVr6FT7VJ7Q9"
    "8z1vPNgzqb1yoN88RlwKPs4Epj2EMqC94khgPc6Ttj1Zi5E8AM+NPPjyLz1iEF0+tjoFPoGbOT0y5La8ZDXXvawRDr31uCy9"
    "03sIPvqsoD1nZDc8UM1BvgtkID2amz4+WrXePfE9gj0rAyU+dAJFPj2+bT6sjj+9sdSgPVtME71YgbC9/+FePdwUoLzNYgI8"
    "k1QLPpomrz1NczG9qzYQPBwPPT3fmH29jBULPoDnSj6ttMa8+ZonvuvS2D3+THc8wGpAPMZNGz0wYTA+J0oyPii6CbwgLv68"
    "W2lIPcCHOz6UhCo+cxaePY1Qhj0d3vI8FFYJPSem+L25pOc9jqNQPiY+CD52k4+9Ma4lvad8Yj5YwAK8hx0qvpC8Fb6LaoY+"
    "r4sIPNi6JrxeZh6+hIlXPps7HD7wAcI7r12hvVIHJT1JxRK9KmfZPfh7lb3ADNO7+XXCvX5yWzufjFI+M2EIvR/PRT7YGo48"
    "kVD2u+b9Lj5ABlY+ocrYPWuMzj26eTA+Zojcvf1rkTzPVwY9cTJxPTjtEj7gQwy+FxXuPSqkdj1tICg+SXYkvpXb9D3qBWU+"
    "oWMcvpRXhr4oWAI+/e5DvZgD6jwKVAo+Py2lPXfESz41aqE9FtcvvvXo0z0+/SY+45ouPoGcuj3TLwc9uurFPakEqT1dBbe8"
    "zgYSPpjlJD5exI09AK8xO3pyG70xFE0+MijQPbvfSL6QsSC+4xNzPrYXKL5CYSU+CJ0cvrAfLD1KvOk8Q+IaPnDDgLs5nYg9"
    "ITGfPWu4Pj6djv89qMpXPl3uCD4TLBq+Is0sPWZr8z2BKko+2eskPXzaq7pKDYk5WWekPbz5WT14NIo9i4mjPORbb72t04y7"
    "/0zpvOaANz4haR897sLOvPmX2D2tniE9gQo7PjU8HL7/o449mtGOvHEvy73NZrq9BycBPe7DijzOTcE99SUaPuzAOD5hnaM9"
    "JagePH1ucb4cFZm8hzwHPk8Thz1+L4+9KV4jPCK8FD6NI3U9JyxJvgiQPj4CEyI8YmpAPpr/hz2jpxs8qahgPgtvfL0aQUC+"
    "0WSHPVPp3j2wQdm9JBfSPb6EY729pj4+gMw+Pi83tD1dykM9IKFYvC6rOz2Z/KE9BDJDPW5BRj5Mws09lN7tvcuaAj6PdbQ9"
    "ZyIKPtddRD5MsXc8vqXzPaMx6j0G+6e9LqguvZt9FD12Obc96UDlvYZA4z0SMQA+0MgLvRR8sjzpxq094fD1vFNCoz0DPom+"
    "JnQuPo2kJD4LPxO9V6PyvHgt4T0+uRc+9D1TPdNL87xlRl4+SOIHPvbCFr2HTDW+iz8/PuLUNj7GTRA+JGXUPF1dbT7uAuy7"
    "slNkPRvuiL05fso9XSRfPi+w8j1A4t07X+oKPnQDrz3kmGc6Qo1TvtAAPL2PiqI9YqkjPULoQT4eBQS+NhQUPSlCTz7RspO8"
    "yXoGPvB20z34S1s8ie0XvbVM2j0cvzk+yJYKPgOwrL3Q4xE+QjvovBcv0z2QCOw9PfAxPv196zzrUPw9LLQTvQjSw7vlOu49"
    "YCacPLOKVzwdm5w9JdSGPU8fjz3SH6A8CVHku/yo8j0xhik+y+isvZ4GlDw2yFE+1ndEvtbniL3XBO091wrqPCzbHD0nNlW9"
    "kOJFPQ6Zhj0Uywa+ShI5vttkMz7i+wg+5EZAPt3A3b3PoSc+GsFKPs6XDT5XaSS+XU2zu+lqaz3SFUc+cHAzPBMPPT4+WJY8"
    "ZtWOvTd+Yr1QVZU7LYttPdFpZj1sLis9GS0Kvsd+Fj4hVr48MNQEvZ/nGr2KVqg5jXHoPa1YAD6Xhq+8UEomPXvGFD5u1zm+"
    "NTJEPSrHbz2tlOG7C92mu5c8Yz5E7wk+ZWUpPm20uj3HDCU+guYMPkhbtz3CX7U8v5UqPprwCz6h5Bg9COKxvPIkrboecAa8"
    "ApdQPu36dr7VY4u8Jo92vFvDO77oksO9KpqbPY1wn71A+SI8G2LtvMzOAT4AvzI+srMZvTidPr7Ahxw+iKqWPbGN0Tz7Qbq9"
    "OKRQPpVdAD428sS8M/kUvLPsYj4kDCU+bv6aO3BJ37zBwxs+WvNAPq6GCr1Dq6K80+4DvpD/MD6XDp49h5nUPcXs572gWxY+"
    "TIQfPnsP/D2fNHy9XchIvZFAJz3IKDM94HZ+PPB0oz3+UJE9dxnyvDBqVT3tYjY+fMyxPXDZ8T0mogM+4ypiPdiRDj5i+788"
    "3/wCPvQhT7zEB2u9rAkdvl3/qTwL8cg9eAMmPnB1ML4RZK488vUtPrIGoDzugJO9J1WNPfqBL7xb97e9grlUvpAzJD1Mv6q9"
    "ro3wPRUh6zsRljY+LU/aPJEW2L0tx829o+gaPqb7HT6sXlW9STwEvbpkLT02jgA+pfg0Pd+H37wsYGg+B7t0PUeJRj4o3LE9"
    "DEiUPX6vkbw6zZy9GYCRvRfPbT22LFI+gI0ovWUlcTz1zA2+ZxQcPQbBsD0WUnU9x5TwPSOzBb4Lk+k9PZQvPklohr2vswo9"
    "lK84uwD6vb2Z9kc+UATbPQeANT7bsEM+f5dMPtKpF7xF4G49JPwGPTFfqj24l4E9IBDXPZowsj2S5ww9ycqKu6WliT1K4nS8"
    "UmsPPiHrAz7y+/k9Ip/SvScFHT5orzU+qK7AvRAbVb4LkM0936t+PMpmt73Tsyg9uPZHPtjzIj0mZIo5clatvZpZIz4VZwg+"
    "ZcUgvQfyGL3TJZc94VTIPZL0cL1SGpU8C3EJPmQ0jjy04SM+gLTOO+JUMT6SH789v2zPO8OCYb5E2xe+lhcNPrytnD1XwTU7"
    "grdmPO7Ksjxlnlu80nowvIi+f73D8QC+DGQEPJ/JQj137tY9+6Q6PhZv1zxCcu+9I/5DPfHpw7wKMjA+kQFLPml8obw2xsi8"
    "ChWvPIHtYTwBJ6Q9sXLgvLB1jT3S8du8RUE7PP8Dgr2K3SA+tFidPS20nj03Z+k8wKonPs/Tqb0NJi894PJfPnE4Ub6GcBi+"
    "nG31PQGqRb3I7h29Q7yruoKrNT2WAFs+3RSAPbEbhb5kXiG55tcfPrfvVjyVgoa9W28uvCvYNz4x9qc9kRqfvWwIlz3Ozae8"
    "Ah+wPXy8Qz2esYE8B9hLPsc+GD2/lgu+J0OfPD0DUz6ymhi++nX/PUfZFb4eKKc9EcYvPY3JAz4aD8i93/gcO/Guj73JufI9"
    "dMeaPeXwYj0OI4k9h+4GO/CfIT43p/283mx2PbvfTz6QLPU992buPfgQHT6at9Q73gJWPXYugT1mtMA9ozpNvbk2HT2PPqI9"
    "NDJWPBDCtz1KLiQ+CVW4PbvDpTv2QMK951ktPm0DFj6WNlU5pOUqvq0z0r2YTqm8YOaCO2JcOj5HXgw+uBSfPaGDGjwucwm+"
    "KBJbOw1szT2gHJ89NXPpPQUAZj6wsQg+jyU1vUfQRT2QI2c+F2qXPQGoyD0Ku+c9gKkJPOlCHz5N0wQ9zuwdu0+wkbt5aIY+"
    "NwzKPWlUL7tDYa49lsarPa7CtTwBinM9zKXuO7pIrr0MuG+9aQ6Su+o2az0NvVM+jz7TPUf/N72B5kY+onHbvCBTBzwsCxA+"
    "XKuMvLtWXD64NiM+PPc1O3X2SD42V+M9ikzXPWwyoz3BjUE9a6wqPo8XlT1lvNq7kR7bPZorarzF0gY8lOI/vAhVqbzjei0+"
    "FIO9PB0UTb44v9w7/SXVPLDsvT1QDAw9vWtgPu49ErzR2Lk9VyomvavluzzezN89zsPcvBWJfD3TaiE+cP8kPRguET5iba69"
    "VIapPflzVD4sFeg9IAqJPUU1/z0AWEs8bbgwPApvHT3mW/m9SSIcPlc9sT3yXsM9njWTvE5pGz5UT5u8J4YQPh+v+T0qlxi9"
    "E/ShO+MkLT0Vhyw+bMOpPcoqij0/Hki+uT3iPaThHT6JKv494iOjPVxl0z0io04+I4BGPlU/7Tz8LW+93YbkPdyePr20RLE9"
    "PCMXPU6g1jwvUpy9qFqavEjibT32jd29SolnPV9aCT3CYgU+G6M3PryUq7vqx0s8CFinvVqAij04TrU8avBCPt6rAz70rkA+"
    "n04WPsE8oD3I5wg9OZd1O2NvTL1BRoG9xvESPvnfub0VygY8BkmOPYpbUj5uV989HgAZPqAwG7zqjG+8DQfLPFTJPL3Aupk9"
    "DkuqvVbSoTwOl+I9FUsevTswm733TQI+6IZaPcVnWb2z37+9qOfCvFo7WL1laI09vGapvLL3UD7rckI9+6uCvI8OCju1p8G8"
    "7w24Pf5xyb26I8O8D3sTPmL49z0NeBo+05omPj0B5jx6Dc69tCaiPQ+uqj25oDE+5fZfvVdtFbrnsym9BRORPadWsbz8SYG+"
    "ZFGOPb1VbD4ftG69O4Tzvf/FCz5QYRk9DvGOvVY5Gz4nW0Q+LeBVPv2PBz4ySSa58Y7KPJE2LD5dtQo8MhsUPNZuXz6kwFy8"
    "5urZPdY2Hr6w8Ns9M6bOPd7IID2wpC48STk9vAAgWD5M5g29p0RxupIHqb0KUS49QN2oPRBLQj1hMFW9HqXxPWWTsz0i55c9"
    "Qrb6uw90SL0d1PQ9FiP7PenzHT7OQhQ+QQiYuwdX373IcUk9NhMuPTQNOz21D1I9I1YLPvHtFT5QSwcIN6xuNwBAAAAAQAAA"
    "UEsDBAAACAgAAAAAAAAAAAAAAAAAAAAAAAALAAcAYmVzdC9kYXRhLzRGQgMAWlpa7ZDwPauC7r200NQ9GI2kvMoo+r09aLc9"
    "TcqMvfFLG73lt2I9iqULPcGttD08oLK9z0O4PW4StT3T7a09WLDXPKs3HT0qbik95xP3PQzOSbyrOcy9wx7Vvadi3zxgW7a9"
    "AxPpu7jhPLxyTuA8yNqgvQdEnj2FQ9C9mxqMvUL7Kj0H+fG9ZnwePeLbKr24Qn69Xx3JPXFJFL3fyDg9VXSovAJZgz2FdQ4+"
    "9aABPrDamr2Qef69oMERvVwnVD39XXc9SgKwPeVMfz0tDRM+n2O8vWBHyL1Cuhi9WeXbPTn4sr0/RdQ9ZUPMPEK/tT0L6gE+"
    "ew8TPoRYET7bpqW7JW0PvVBLBwgQJzRaAAEAAAABAABQSwMEAAAICAAAAAAAAAAAAAAAAAAAAAAAAAsABwBiZXN0L2RhdGEv"
    "NUZCAwBaWlrmsMy8umWqvX1UN71q+am9AAAAAJ25R72EC4e9YoMuvZjNgb1EwHS8HmEwvQAAAACoixe9AUcmvQqUJb2d7Jc9"
    "cp0Ovd4HQb2e9jC9d4qWPRn6rL0ZCou9i21evfh01L2I5Ee9uKsQPRJkr72x+IO9bUonvYh4yL1Zt3i9TwmivXm8hL1vhKO9"
    "hY+IvSEPP73vWiK9VQyNvfA0OL3XQRY9U1x9vTwbTLxP4Qy9AAAAAAAAAADDfIE9/P5qvbpaZb0MviW9HUk6vXcdHb0dCb+9"
    "KSNJvd0ng73+cQy96cCvvbL0PL2wbYe9lFYgvfPMhb3nNEK9HGYDvWLap737Yku9TR42vXOpiL1b1Wy9Fc2IvQAAAACdAHW9"
    "XLOLvc/1QL3XuYi987dGveaWXb0AAAAABtk+vVa1Vb1xX0y9X497PdXbSb3svzO9tzBXvUxhaj09A469AyCFvSacU711cIi9"
    "Xb5EvT2k3zy4AIm9JWSDvdnfWL3aZ4y9kNIivaWyhr1dOIC9Y5eSvcU4dL3BgV+9v7pWvb7pd72lE1C9PStXu0TAgL2vJzq9"
    "B6hZvQAAAAAAAAAAc2o4PQp0X71/WGu96903vTdFZL026UC989aWvZ5jZ73UB4C9BEpBvcI2aL3qM2C9eB+ZvdEoUb2vZo29"
    "Hd9VvTycNL32WJe83+NdvVHtIL3GGoC9GThLvZ1zh70AAAAAokJhvRzVeL2oo1S9LXUvvb3ntbxvymy9AAAAAJ/YQL1RDFm9"
    "LeVUvfsPoD2vwEC9nqJBvai1X714I4k9J7Onvcryhb1ZPie9t9GbvXE+SL1zQrw8LcxsvV9Llr35Uly9WrSIva1OSb3bTZe9"
    "zfOBvd/of7105IC9bM5svSNhRL2qfI29yXs3veSfn7xnlYO97YjlvOUKGL0AAAAAAAAAAJZC0Txiy4S9Gsdtvb71KL2NLVi9"
    "LP9JvUcAKL2L0Va9BSaLvaJHSr3rGJ+9F/1wvSHlkr0HHEy9WeSCvdp6Tb3SPCG9XoTLvGu3Qb0mWii9WDyDvSNHe71RT4e9"
    "AAAAALByb72OMou9b885vYuPkr3dRmO9S0RLvQAAAABHezq91RVkvTmGO70bjxQ98+ZNvX5FZ72BAkS9ZNSJPcw2e72BpoS9"
    "JyNEvY1aer3n/Fm9thkBPZQpcr0CvIS9maBFvfUvk733RlO9c4eOvSyAfL0A1IK9NYNzvR6LVb0RsUu9qZtkvT9EYb1QfT89"
    "ps9vvW7VN73HT0m9AAAAAAAAAACdhyk9dF05vZ9oZ73DX0K9ZSFdvWqGP72vrYS94JxovZIqcr2MIjq9ztJivSvtVb1nYYi9"
    "YmtDvWS9hr3D2lm9yV9AvcVNrbx9jm69mH/7vAAvgr0zS3u9mIiNvQAAAABLj4W9sd6NvSiFMr35VZm9R4ZivaD4UL0AAAAA"
    "BYgdvSFFWb0+2Ue99wWyPY7EOb1hSoK9DRVUvVfNtD2OpnS9eHGPvWKNO71LxL29OrpcvQtfTz0lYa+9pDl8vceLN72zuLS9"
    "VbY3vcDKmL2g1Iq9IKmQveTVcL0FL0q9nO9NvWoxZL2ctmC95f42PeeYh723/nq8eutQvQAAAAAAAAAABscqPf0tUb1LUoq9"
    "F048vazDX71LBkS9UnuRvYCta721xoi905MwvfIHNr3JbGa9AtqXvdzENr1BgIi9SZtavZ5oLb059nm9vEFrvQOq4rwBf4m9"
    "HXZ8vTmBjL0AAAAAZ8iKvdbunb3pgCK9hraYve1wl72aqlW9AAAAAOpHIr1fxT69eSFEva2KqD2SOzO9TZNSvS/3Tb3Eg789"
    "J6uLvY41l716Tye9Gqe+veTiN711B049KRaxvfREoL01zFu9XOHKvb65YL2xoqe9/3GNvZpxnL22rF+9E4xDvd5xS71DFYa9"
    "lqxAvfn0cT30Zou9Lf34vJQrVr0AAAAAAAAAAG1Jqz0R0j+9UweAvRUTH710QGK9iPs8vTpdor3tkIC9ZemFveOjPb1UFoK9"
    "+nRUvYVjpr2wzy29IC+JvYOZYb19GyK90lWDvbFfVr3qASy9DAl7vfEFe72uC3y9AAAAAGTqkL0Cw5K9dDdMvcrhhr3fxDq9"
    "Idx5vQAAAACGFy+9vJBXvW2EXL1ocd09wXtCvUVSS730CXu9LMiePQcfir2hAI297LYavUbxoL0lIjG9eX4TPaPOh72fnpm9"
    "LutpvZL6kL1Sw3m96s2evQtAjb0WPli9M+V3vWDpdL0G0FG9hG6DveTsQr0CIl49jHaTvWQQCr0j9na9AAAAAAAAAABRd449"
    "0gB1vUMrZr3JzDy9UKpxvQbEV70k52G9t/ZivWf8ir0UY069yfCDvYenUb0Ee5O9fvw6vbUAlL1aqWe9H34yvTzf0bvRxTa9"
    "UEsHCAHh1hIABwAAAAcAAFBLAwQAAAgIAAAAAAAAAAAAAAAAAAAAAAAACwAHAGJlc3QvZGF0YS82RkIDAFpaWouWhbynafK8"
    "QBetvMI2zrzIB0e8iYTOu+OjtLxQSwcIBHiFsBwAAAAcAAAAUEsDBAAACAgAAAAAAAAAAAAAAAAAAAAAAAAMACoAYmVzdC92"
    "ZXJzaW9uRkImAFpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaMwpQSwcI0Z5nVQIAAAACAAAAUEsDBAAA"
    "CAgAAAAAAAAAAAAAAAAAAAAAAAAbADUAYmVzdC8uZGF0YS9zZXJpYWxpemF0aW9uX2lkRkIxAFpaWlpaWlpaWlpaWlpaWlpa"
    "WlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlpaWlowNzIwNzQ1MDEwMDc1Mzk2NTg1NjA0MjczODc3MjAwOTM4OTg1OTgy"
    "UEsHCE8HNfsoAAAAKAAAAFBLAQIAAAAACAgAAAAAAAB/c8kyZgYAAGYGAAANAAAAAAAAAAAAAAAAAAAAAABiZXN0L2RhdGEu"
    "cGtsUEsBAgAAAAAICAAAAAAAALfv3IMBAAAAAQAAABQAAAAAAAAAAAAAAAAAtgYAAGJlc3QvLmZvcm1hdF92ZXJzaW9uUEsB"
    "AgAAAAAICAAAAAAAAD93cekCAAAAAgAAABcAAAAAAAAAAAAAAAAAEQcAAGJlc3QvLnN0b3JhZ2VfYWxpZ25tZW50UEsBAgAA"
    "AAAICAAAAAAAAIU94xkGAAAABgAAAA4AAAAAAAAAAAAAAAAAkgcAAGJlc3QvYnl0ZW9yZGVyUEsBAgAAAAAICAAAAAAAABzf"
    "RCEEAAAABAAAAAsAAAAAAAAAAAAAAAAAFggAAGJlc3QvZGF0YS8wUEsBAgAAAAAICAAAAAAAAEAtEzQAGQAAABkAAAsAAAAA"
    "AAAAAAAAAAAAlAgAAGJlc3QvZGF0YS8xUEsBAgAAAAAICAAAAAAAAH4D9z8AAQAAAAEAAAsAAAAAAAAAAAAAAAAAECIAAGJl"
    "c3QvZGF0YS8yUEsBAgAAAAAICAAAAAAAADesbjcAQAAAAEAAAAsAAAAAAAAAAAAAAAAAUCMAAGJlc3QvZGF0YS8zUEsBAgAA"
    "AAAICAAAAAAAABAnNFoAAQAAAAEAAAsAAAAAAAAAAAAAAAAAkGMAAGJlc3QvZGF0YS80UEsBAgAAAAAICAAAAAAAAAHh1hIA"
    "BwAAAAcAAAsAAAAAAAAAAAAAAAAA0GQAAGJlc3QvZGF0YS81UEsBAgAAAAAICAAAAAAAAAR4hbAcAAAAHAAAAAsAAAAAAAAA"
    "AAAAAAAAEGwAAGJlc3QvZGF0YS82UEsBAgAAAAAICAAAAAAAANGeZ1UCAAAAAgAAAAwAAAAAAAAAAAAAAAAAbGwAAGJlc3Qv"
    "dmVyc2lvblBLAQIAAAAACAgAAAAAAABPBzX7KAAAACgAAAAbAAAAAAAAAAAAAAAAANJsAABiZXN0Ly5kYXRhL3NlcmlhbGl6"
    "YXRpb25faWRQSwYGLAAAAAAAAAAeAy0AAAAAAAAAAAANAAAAAAAAAA0AAAAAAAAAEAMAAAAAAAB4bQAAAAAAAFBLBgcAAAAA"
    "iHAAAAAAAAABAAAAUEsFBgAAAAANAA0AEAMAAHhtAAAAAA=="
)


def load_p3_checkpoint():
    """Verify the exact artifact and load only tensors and primitive metadata."""
    raw = base64.b64decode(CHECKPOINT_BASE64, validate=True)
    git_header = f"blob {len(raw)}\0".encode("ascii")
    if hashlib.sha1(git_header + raw).hexdigest() != CHECKPOINT_GIT_SHA:
        raise ValueError("Embedded P3 checkpoint integrity mismatch")
    state = torch.load(io.BytesIO(raw), map_location="cpu", weights_only=True)
    if state.get("signature") != signature():
        raise ValueError("P3 checkpoint architecture/profile signature mismatch")
    interval = state.get("interval")
    if type(interval) is not int or interval < 1:
        raise ValueError("Invalid P3 decision interval")
    if not math.isfinite(state["validation_loss"]):
        raise ValueError("Nonfinite checkpoint validation loss")
    # Match training-time model construction without changing global RNG state.
    with torch.random.fork_rng(devices=[]):
        policy = QPolicy()
    policy.load_state_dict(state["model"], strict=True)
    policy.validate()
    policy.epsilon.zero_()
    policy.eval()
    policy.requires_grad_(False)
    return policy, interval


class TransferPolicy(smart_buildings_inference_server.Policy):
    """P3 chooses a V23 profile; V23 still dispatches every five minutes."""

    def __init__(self, building_meta_data, action_spec):
        super().__init__(building_meta_data, action_spec)
        self.baseline = Version23Policy(building_meta_data, action_spec)
        self.controller = ProfileController(self.baseline)
        self.model, self.interval = load_p3_checkpoint()
        self.reset()

    def reset(self):
        """Restart the macro-step clock and previous-profile feature per episode."""
        if hasattr(self, "controller"):
            self.baseline.reset()
            self.controller.select(0)
        self.step = 0

    def get_action(self, obs):
        # Identical timing and feature construction to common.rollout(eval).
        if self.step % self.interval == 0:
            features = self.controller.features(obs)
            with torch.no_grad():
                logits, _ = self.model(torch.from_numpy(features))
            self.controller.select(int(logits.argmax().item()))
        command = self.controller.get_action(obs)
        self.step += 1
        return command


# Official Kaggle gateway; exactly one runnable notebook cell.
smart_buildings_inference_server.SmartBuildingsInferenceServer(TransferPolicy).run()
```
