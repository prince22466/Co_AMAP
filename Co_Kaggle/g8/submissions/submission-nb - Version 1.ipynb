NeurIPS 2026 Smart Buildings — minimal valid policy submission
This notebook implements the organizer-published minimal policy interface, for an initial submission-format test. It is deliberately a zero-action baseline, not a competitive controller or an RL model. The competition gateway owns the hidden building simulations and creates the scored output.
- Runtime: CPU (no GPU required), internet OFF.
- No external model files or starter-kit installation required for this reference policy.
- Do not add an explicit 8,640-step rollout or write your own submission.parquet.
- The kaggle_evaluation package and hidden-scenario gateway must be provided by Kaggle's competition environment.
Source: organizer's submission specification and the Kaggle competition.

```python
from typing import Any, Dict

import numpy as np
from kaggle_evaluation import smart_buildings_inference_server


class TransferPolicy(smart_buildings_inference_server.Policy):
    """Minimal, valid reference control policy (not optimized)."""

    def __init__(self, building_meta_data, action_spec):
        super().__init__(building_meta_data, action_spec)
        self.n_discrete = int(building_meta_data["n_discrete_action"])
        self.n_continuous = int(building_meta_data["n_continuous_action"])

    def get_action(self, obs: np.ndarray) -> Dict[str, Any]:
        return {
            "continuous_action": np.zeros(self.n_continuous, dtype=np.float32),
            "discrete_action": np.zeros(self.n_discrete, dtype=np.int32),
        }
```

```python
# The competition gateway handles all rollouts and generates submission.parquet.
smart_buildings_inference_server.SmartBuildingsInferenceServer(TransferPolicy).run()
```
